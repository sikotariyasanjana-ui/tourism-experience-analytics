## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# ============================================================
# STEP 13: DATA WRANGLING
# Tourism Experience Analytics Project
# ============================================================

import pandas as pd

print("=" * 80)
print("DATA WRANGLING")
print("=" * 80)

# ------------------------------------------------------------
# 1. Create copies of original datasets
# ------------------------------------------------------------

transaction = transaction_df.copy()
user = user_df.copy()
city = city_df.copy()
country = country_df.copy()
region = region_df.copy()
continent = continent_df.copy()
item = item_df.copy()
type_df_w = type_df.copy()
mode = mode_df.copy()


# ------------------------------------------------------------
# 2. Merge Transaction with User
# Relationship: Transaction.UserId → User.UserId
# ------------------------------------------------------------

merged_df = transaction.merge(
    user,
    on="UserId",
    how="left",
    suffixes=("", "_User")
)

print(f"After Transaction + User : {merged_df.shape}")


# ------------------------------------------------------------
# 3. Merge User information with City
# Relationship: User.CityId → City.CityId
# ------------------------------------------------------------

merged_df = merged_df.merge(
    city,
    on="CityId",
    how="left",
    suffixes=("", "_City")
)

print(f"After + City             : {merged_df.shape}")


# ------------------------------------------------------------
# 4. Merge City with Country
# Relationship: City.CountryId → Country.CountryId
# ------------------------------------------------------------

merged_df = merged_df.merge(
    country,
    on="CountryId",
    how="left",
    suffixes=("", "_Country")
)

print(f"After + Country          : {merged_df.shape}")


# ------------------------------------------------------------
# 5. Merge Country with Region
# Relationship: Country.RegionId → Region.RegionId
# ------------------------------------------------------------

merged_df = merged_df.merge(
    region,
    on="RegionId",
    how="left",
    suffixes=("", "_Region")
)

print(f"After + Region           : {merged_df.shape}")


# ------------------------------------------------------------
# 6. Merge Region with Continent
# Relationship: Region.ContinentId → Continent.ContinentId
# ------------------------------------------------------------

merged_df = merged_df.merge(
    continent,
    on="ContinentId",
    how="left",
    suffixes=("", "_Continent")
)

print(f"After + Continent        : {merged_df.shape}")


# ------------------------------------------------------------
# 7. Merge Transaction with Item/Attraction
# Relationship: Transaction.AttractionId → Item.AttractionId
# ------------------------------------------------------------

merged_df = merged_df.merge(
    item,
    on="AttractionId",
    how="left",
    suffixes=("", "_Item")
)

print(f"After + Item             : {merged_df.shape}")


# ------------------------------------------------------------
# 8. Merge Attraction with Type
# Relationship: Item.AttractionTypeId → Type.AttractionTypeId
# ------------------------------------------------------------

merged_df = merged_df.merge(
    type_df_w,
    on="AttractionTypeId",
    how="left",
    suffixes=("", "_Type")
)

print(f"After + Attraction Type   : {merged_df.shape}")


# ------------------------------------------------------------
# 9. Merge Transaction with Visit Mode
# Relationship: Transaction.VisitMode → Mode.VisitModeId
# ------------------------------------------------------------

merged_df = merged_df.merge(
    mode,
    left_on="VisitMode",
    right_on="VisitModeId",
    how="left",
    suffixes=("", "_Mode")
)

print(f"After + Visit Mode        : {merged_df.shape}")


# ------------------------------------------------------------
# 10. Remove completely duplicate rows
# ------------------------------------------------------------

before_duplicates = len(merged_df)

merged_df = merged_df.drop_duplicates().reset_index(drop=True)

after_duplicates = len(merged_df)

print("\nDuplicate rows removed:", before_duplicates - after_duplicates)


# ------------------------------------------------------------
# 11. Display final dataset information
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("FINAL WRANGLED DATASET")
print("=" * 80)

print("Rows    :", merged_df.shape[0])
print("Columns :", merged_df.shape[1])

print("\nFirst 5 Rows:")
display(merged_df.head())


# ------------------------------------------------------------
# 12. Check missing values after merging
# ------------------------------------------------------------

print("\nMissing Values After Data Wrangling:")

missing_after_merge = merged_df.isnull().sum()

missing_after_merge = missing_after_merge[
    missing_after_merge > 0
].sort_values(ascending=False)

if missing_after_merge.empty:
    print("✅ No missing values found after merging.")
else:
    display(missing_after_merge.to_frame("Missing Values"))


print("\n" + "=" * 80)
print("✅ DATA WRANGLING COMPLETED SUCCESSFULLY!")
print("=" * 80)# Write your code to make your dataset analysis ready.
````

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ------------------------------------------------------------
# 1. Create a safe analysis dataframe
# ------------------------------------------------------------

# Start from Transaction data
viz_df = transaction_df.copy()

# Merge User information
viz_df = viz_df.merge(
    user_df,
    on="UserId",
    how="left"
)

# Merge Mode information
viz_df = viz_df.merge(
    mode_df,
    left_on="VisitMode",
    right_on="VisitModeId",
    how="left"
)

# Merge Attraction information
# Only use non-null unique AttractionId records
item_clean = item_df.dropna(subset=["AttractionId"]).drop_duplicates(
    subset=["AttractionId"]
)

viz_df = viz_df.merge(
    item_clean,
    on="AttractionId",
    how="left"
)

# Merge Attraction Type
viz_df = viz_df.merge(
    type_df,
    on="AttractionTypeId",
    how="left"
)

# Merge City
city_clean = city_df.drop_duplicates(subset=["CityId"])

viz_df = viz_df.merge(
    city_clean[["CityId", "CityName"]],
    on="CityId",
    how="left"
)

# Merge Country
country_clean = country_df.drop_duplicates(subset=["CountryId"])

viz_df = viz_df.merge(
    country_clean[["CountryId", "Country"]],
    on="CountryId",
    how="left"
)

# Merge Region
region_clean = (
    region_df.dropna(subset=["RegionId"])
             .drop_duplicates(subset=["RegionId"])
)

viz_df = viz_df.merge(
    region_clean[["RegionId", "Region"]],
    on="RegionId",
    how="left"
)

# Merge Continent
continent_clean = (
    continent_df.dropna(subset=["ContinentId"])
                .drop_duplicates(subset=["ContinentId"])
)

viz_df = viz_df.merge(
    continent_clean[["ContinentId", "Continent"]],
    on="ContinentId",
    how="left"
)

# Remove accidental duplicate rows after merging
viz_df = viz_df.drop_duplicates().reset_index(drop=True)

# Remove placeholder "-" values where useful
for col in ["VisitMode", "VisitMode_y", "AttractionType", "Attraction",
            "CityName", "Country", "Region", "Continent"]:
    if col in viz_df.columns:
        viz_df[col] = viz_df[col].replace("-", np.nan)

print("✅ Visualization dataset created successfully!")
print("Shape:", viz_df.shape)


# ============================================================
# CHART 1: Rating Distribution
# ============================================================

plt.figure(figsize=(10, 5))
sns.histplot(viz_df["Rating"].dropna(), bins=5, kde=True)
plt.title("1. Distribution of Attraction Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Visits")
plt.tight_layout()
plt.show()# Chart - 1 visualization code
````

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

H₀ (Null Hypothesis) : There is no significant difference in average attraction ratings between the selected visit-mode groups.
H₁ (Alternative Hypothesis) : There is a significant difference in average attraction ratings between the selected visit-mode groups.

#### 2. Perform an appropriate statistical test.

````python
# ============================================================
# STEP 17: STATISTICAL TESTING - P-VALUE
# Tourism Experience Analytics Project
# ============================================================

import pandas as pd
from scipy.stats import ttest_ind, pearsonr

# Significance level
alpha = 0.05

print("=" * 90)
print("STATISTICAL TESTING AND P-VALUE")
print("=" * 90)


# ============================================================
# HYPOTHESIS 1: Visit Mode vs Rating
# Independent Two-Sample T-Test
# ============================================================

print("\n" + "-" * 90)
print("HYPOTHESIS 1: VISIT MODE vs RATING")
print("-" * 90)

# Find the visit mode column
if "VisitMode_y" in viz_df.columns:
    mode_col = "VisitMode_y"
elif "VisitMode" in viz_df.columns:
    mode_col = "VisitMode"
else:
    mode_col = None

if mode_col is not None:

    # Find two most common visit modes
    mode_counts = viz_df[mode_col].dropna().value_counts()

    if len(mode_counts) >= 2:

        group1_name = mode_counts.index[0]
        group2_name = mode_counts.index[1]

        group1 = pd.to_numeric(
            viz_df.loc[viz_df[mode_col] == group1_name, "Rating"],
            errors="coerce"
        ).dropna()

        group2 = pd.to_numeric(
            viz_df.loc[viz_df[mode_col] == group2_name, "Rating"],
            errors="coerce"
        ).dropna()

        # Statistical test
        t_stat_1, p_value_1 = ttest_ind(
            group1,
            group2,
            equal_var=False
        )

        print(f"Group 1       : {group1_name}")
        print(f"Group 2       : {group2_name}")
        print(f"Mean Group 1  : {group1.mean():.4f}")
        print(f"Mean Group 2  : {group2.mean():.4f}")
        print(f"T-statistic   : {t_stat_1:.4f}")
        print(f"P-value       : {p_value_1:.6f}")

        if p_value_1 < alpha:
            print("Decision      : Reject H0")
            print("Conclusion    : Significant difference in ratings.")
        else:
            print("Decision      : Fail to Reject H0")
            print("Conclusion    : No significant difference in ratings.")

    else:
        print("⚠️ Not enough visit-mode groups.")

else:
    print("⚠️ VisitMode column not found.")



````

##### Which statistical test have you done to obtain P-Value?

Independent Two-Sample t-Test

##### Why did you choose the specific statistical test?

We used an Independent Two-Sample t-Test because we wanted to compare the average ratings of two independent visit-mode groups. The test determines whether the difference between their mean ratings is statistically significant.

### 1. Handling Missing Values

````python
# ============================================================
# STEP 18: HANDLING MISSING VALUES & MISSING VALUE IMPUTATION
# Tourism Experience Analytics Project
# ============================================================

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. Dataset path
# ------------------------------------------------------------

data_path = r"C:\Users\Admin\OneDrive\Desktop\TOURISM_PROJECT\DATA"


# ------------------------------------------------------------
# 2. Load all datasets
# ------------------------------------------------------------

city_df = pd.read_csv(data_path + r"\City.csv", encoding="latin1")
continent_df = pd.read_csv(data_path + r"\Continent.csv", encoding="latin1")
country_df = pd.read_csv(data_path + r"\Country.csv", encoding="latin1")
item_df = pd.read_csv(data_path + r"\Item.csv", encoding="latin1")
mode_df = pd.read_csv(data_path + r"\Mode.csv", encoding="latin1")
region_df = pd.read_csv(data_path + r"\Region.csv", encoding="latin1")
transaction_df = pd.read_csv(data_path + r"\Transaction.csv", encoding="latin1")
type_df = pd.read_csv(data_path + r"\Type.csv", encoding="latin1")
user_df = pd.read_csv(data_path + r"\User.csv", encoding="latin1")


# ------------------------------------------------------------
# 3. Store datasets
# ------------------------------------------------------------

datasets = {
    "City": city_df,
    "Continent": continent_df,
    "Country": country_df,
    "Item": item_df,
    "Mode": mode_df,
    "Region": region_df,
    "Transaction": transaction_df,
    "Type": type_df,
    "User": user_df
}


# ------------------------------------------------------------
# 4. Create copies
# ------------------------------------------------------------

cleaned_datasets = {
    name: df.copy()
    for name, df in datasets.items()
}


# ------------------------------------------------------------
# 5. Check missing values BEFORE imputation
# ------------------------------------------------------------

print("=" * 90)
print("MISSING VALUES BEFORE IMPUTATION")
print("=" * 90)

for name, df in cleaned_datasets.items():

    missing = df.isnull().sum()
    missing = missing[missing > 0]

    print(f"\n{name.upper()} DATASET")
    print("-" * 60)

    if missing.empty:
        print("✅ No missing values found.")
    else:
        print(missing)


# ------------------------------------------------------------
# 6. Perform missing value imputation
# ------------------------------------------------------------

for name, df in cleaned_datasets.items():

    for column in df.columns:

        # Check whether missing values exist
        if df[column].isnull().sum() == 0:
            continue

        # Numerical columns → Median
        if pd.api.types.is_numeric_dtype(df[column]):

            median_value = df[column].median()

            if pd.isna(median_value):
                df[column] = df[column].fillna(0)
            else:
                df[column] = df[column].fillna(median_value)

        # Categorical columns → Mode
        else:

            mode_value = df[column].mode()

            if not mode_value.empty:
                df[column] = df[column].fillna(mode_value.iloc[0])
            else:
                df[column] = df[column].fillna("Unknown")


# ------------------------------------------------------------
# 7. Check missing values AFTER imputation
# ------------------------------------------------------------

print("\n" + "=" * 90)
print("MISSING VALUES AFTER IMPUTATION")
print("=" * 90)

total_remaining = 0

for name, df in cleaned_datasets.items():

    missing = df.isnull().sum()
    missing = missing[missing > 0]

    print(f"\n{name.upper()} DATASET")
    print("-" * 60)

    if missing.empty:
        print("✅ No missing values remain.")
    else:
        print(missing)
        total_remaining += missing.sum()


# ------------------------------------------------------------
# 8. Create cleaned dataset variables
# ------------------------------------------------------------

city_clean = cleaned_datasets["City"]
continent_clean = cleaned_datasets["Continent"]
country_clean = cleaned_datasets["Country"]
item_clean = cleaned_datasets["Item"]
mode_clean = cleaned_datasets["Mode"]
region_clean = cleaned_datasets["Region"]
transaction_clean = cleaned_datasets["Transaction"]
type_clean = cleaned_datasets["Type"]
user_clean = cleaned_datasets["User"]


# ------------------------------------------------------------
# 9. Final result
# ------------------------------------------------------------

print("\n" + "=" * 90)
print("FINAL RESULT")
print("=" * 90)

print("Total remaining missing values:", total_remaining)

if total_remaining == 0:
    print("✅ Missing value handling and imputation completed successfully!")
else:
    print("⚠️ Some missing values still remain.")

print("=" * 90)# Handling Missing Values & Missing Value Imputation
````

#### What all missing value imputation techniques have you used and why did you use those techniques?

. In our Tourism project, we used median imputation for numerical variables and mode imputation for categorical variables. We selected median            imputation because it is less sensitive to outliers and provides a reliable central value for numerical data. 
. We selected mode imputation because categorical variables contain discrete categories, and replacing missing values with the most frequently           occurring category preserves the existing category distribution. For a completely empty column,
. we used a fallback value such as 0 for numerical data or ‘Unknown’ for categorical data. These techniques were used to handle missing values while     retaining as much information as possible for further EDA and machine-learning tasks.

### 2. Handling Outliers

````python
# ============================================================
# STEP 19: HANDLING OUTLIERS - CORRECTED VERSION
# Tourism Experience Analytics Project
# ============================================================

import pandas as pd
import numpy as np

print("=" * 90)
print("OUTLIER DETECTION AND HANDLING")
print("=" * 90)

# ------------------------------------------------------------
# 1. Create a copy of the cleaned Transaction dataset
# ------------------------------------------------------------

outlier_df = transaction_clean.copy()


# ------------------------------------------------------------
# 2. Make Rating numeric
# ------------------------------------------------------------

if "Rating" in outlier_df.columns:
    outlier_df["Rating"] = pd.to_numeric(
        outlier_df["Rating"],
        errors="coerce"
    )


# ------------------------------------------------------------
# 3. Check Rating outliers using IQR
# ------------------------------------------------------------

if "Rating" in outlier_df.columns:

    Q1 = outlier_df["Rating"].quantile(0.25)
    Q3 = outlier_df["Rating"].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Identify outliers
    outlier_mask = (
        (outlier_df["Rating"] < lower_bound) |
        (outlier_df["Rating"] > upper_bound)
    )

    outlier_count = outlier_mask.sum()

    print("\nRating Outlier Summary")
    print("-" * 60)
    print(f"Q1                  : {Q1:.2f}")
    print(f"Q3                  : {Q3:.2f}")
    print(f"IQR                 : {IQR:.2f}")
    print(f"Lower Bound         : {lower_bound:.2f}")
    print(f"Upper Bound         : {upper_bound:.2f}")
    print(f"Outliers Found      : {outlier_count}")

    # --------------------------------------------------------
    # 4. Handle Rating outliers using capping
    # --------------------------------------------------------

    outlier_df["Rating"] = outlier_df["Rating"].clip(
        lower=lower_bound,
        upper=upper_bound
    )

    print("✅ Rating outliers capped successfully.")


# ------------------------------------------------------------
# 5. Validate VisitYear
# ------------------------------------------------------------

if "VisitYear" in outlier_df.columns:

    outlier_df["VisitYear"] = pd.to_numeric(
        outlier_df["VisitYear"],
        errors="coerce"
    )

    print("\nVisitYear Validation")
    print("-" * 60)
    print("Minimum Year:", outlier_df["VisitYear"].min())
    print("Maximum Year:", outlier_df["VisitYear"].max())

    # We DO NOT apply IQR capping to VisitYear.


# ------------------------------------------------------------
# 6. Validate VisitMonth
# ------------------------------------------------------------

if "VisitMonth" in outlier_df.columns:

    outlier_df["VisitMonth"] = pd.to_numeric(
        outlier_df["VisitMonth"],
        errors="coerce"
    )

    print("\nVisitMonth Validation")
    print("-" * 60)
    print("Minimum Month:", outlier_df["VisitMonth"].min())
    print("Maximum Month:", outlier_df["VisitMonth"].max())

    # Valid calendar months are 1 to 12
    invalid_months = (
        (outlier_df["VisitMonth"] < 1) |
        (outlier_df["VisitMonth"] > 12)
    ).sum()

    print("Invalid Month Values:", invalid_months)

    # Fix only genuinely invalid calendar months
    if invalid_months > 0:
        outlier_df.loc[
            ~outlier_df["VisitMonth"].between(1, 12),
            "VisitMonth"
        ] = np.nan

        # Fill invalid months using the median valid month
        month_median = outlier_df["VisitMonth"].median()

        if pd.notna(month_median):
            outlier_df["VisitMonth"] = outlier_df["VisitMonth"].fillna(
                month_median
            )

        print("✅ Invalid month values handled.")


# ------------------------------------------------------------
# 7. Final summary
# ------------------------------------------------------------

print("\n" + "=" * 90)
print("FINAL OUTLIER-HANDLING SUMMARY")
print("=" * 90)

print("Original Transaction Rows :", len(transaction_clean))
print("Final Transaction Rows    :", len(outlier_df))

if "Rating" in outlier_df.columns:
    print(
        "Rating Minimum            :",
        round(outlier_df["Rating"].min(), 2)
    )
    print(
        "Rating Maximum            :",
        round(outlier_df["Rating"].max(), 2)
    )

print("\n✅ OUTLIER HANDLING COMPLETED SUCCESSFULLY!")
print("=" * 90)
````

  For VisitYear and VisitMonth, we used range validation because they are calendar variables; VisitMonth must lie between 1 and 12, while VisitYear      should be within the valid year range of the dataset.

````python
# ============================================================
# STEP 24: REMOVE URLs & WORDS CONTAINING DIGITS
# Tourism Experience Analytics Project
# ============================================================

import pandas as pd
import re

print("=" * 90)
print("REMOVE URLs & REMOVE WORDS CONTAINING DIGITS")
print("=" * 90)


# ------------------------------------------------------------
# 1. Check whether viz_df exists
# ------------------------------------------------------------

if "viz_df" not in globals():

    print("⚠️ viz_df is not defined.")
    print("Please run your Data Wrangling / Data Preparation cell first.")

else:

    # --------------------------------------------------------
    # 2. Identify text/object columns
    # --------------------------------------------------------

    text_columns = viz_df.select_dtypes(
        include=["object", "string"]
    ).columns.tolist()

    print("Text columns found:")
    print(text_columns)


    # --------------------------------------------------------
    # 3. Function to remove URLs
    # --------------------------------------------------------

    def remove_urls(text):
        """
        Remove HTTP, HTTPS, and WWW URLs from text.
        """

        if pd.isna(text):
            return text

        text = str(text)

        # Remove URLs
        text = re.sub(
            r"https?://\S+|www\.\S+",
            "",
            text,
            flags=re.IGNORECASE
        )

        # Remove extra spaces
        text = " ".join(text.split())

        return text


    # --------------------------------------------------------
    # 4. Function to remove words containing digits
    # --------------------------------------------------------

    def remove_words_with_digits(text):
        """
        Remove complete words that contain one or more digits.
        
        Examples:
        ABC123    -> removed
        A12B      -> removed
        2024      -> removed
        Museum    -> kept
        India     -> kept
        """

        if pd.isna(text):
            return text

        text = str(text)

        # Remove words containing at least one digit
        text = re.sub(
            r"\b\S*\d\S*\b",
            "",
            text
        )

        # Remove extra spaces
        text = " ".join(text.split())

        return text


    # --------------------------------------------------------
    # 5. Apply both cleaning operations
    # --------------------------------------------------------

    for column in text_columns:

        # Remove URLs
        viz_df[column] = viz_df[column].apply(
            remove_urls
        )

        # Remove words containing digits
        viz_df[column] = viz_df[column].apply(
            remove_words_with_digits
        )


    # --------------------------------------------------------
    # 6. Display sample values
    # --------------------------------------------------------

    print("\n" + "=" * 90)
    print("SAMPLE VALUES AFTER CLEANING")
    print("=" * 90)

    for column in text_columns:

        print(f"\nColumn: {column}")

        display(
            viz_df[[column]].head()
        )


    # --------------------------------------------------------
    # 7. Final confirmation
    # --------------------------------------------------------

    print("\n" + "=" * 90)
    print("✅ URLs REMOVED SUCCESSFULLY!")
    print("✅ WORDS CONTAINING DIGITS REMOVED SUCCESSFULLY!")
    print("=" * 90)# Remove URLs & Remove words and digits contain digits
````

#### 8. Text Normalization

````python
# ============================================================
# STEP 30: TEXT NORMALIZATION
# STEMMING AND LEMMATIZATION
# Tourism Experience Analytics Project
# ============================================================

import pandas as pd
import re

# NLTK imports
import nltk
from nltk.stem import PorterStemmer, WordNetLemmatizer

# ------------------------------------------------------------
# 1. Download required NLTK resources
# ------------------------------------------------------------

nltk.download("wordnet")
nltk.download("omw-1.4")

print("=" * 90)
print("TEXT NORMALIZATION: STEMMING & LEMMATIZATION")
print("=" * 90)


# ------------------------------------------------------------
# 2. Check whether viz_df exists
# ------------------------------------------------------------

if "viz_df" not in globals():

    print("⚠️ viz_df is not defined.")
    print("Please run your Data Wrangling / Data Preparation cell first.")

else:

    # --------------------------------------------------------
    # 3. Identify text columns
    # --------------------------------------------------------

    text_columns = viz_df.select_dtypes(
        include=["object", "string"]
    ).columns.tolist()

    print("Text columns found:")
    print(text_columns)


    # --------------------------------------------------------
    # 4. Create Stemmer and Lemmatizer
    # --------------------------------------------------------

    stemmer = PorterStemmer()
    lemmatizer = WordNetLemmatizer()


    # --------------------------------------------------------
    # 5. Stemming function
    # --------------------------------------------------------

    def stemming_text(text):

        if pd.isna(text):
            return text

        text = str(text).lower()

        # Extract words
        words = re.findall(r"\b[a-zA-Z]+\b", text)

        # Apply stemming
        stemmed_words = [
            stemmer.stem(word)
            for word in words
        ]

        return " ".join(stemmed_words)


    # --------------------------------------------------------
    # 6. Lemmatization function
    # --------------------------------------------------------

    def lemmatizing_text(text):

        if pd.isna(text):
            return text

        text = str(text).lower()

        # Extract words
        words = re.findall(r"\b[a-zA-Z]+\b", text)

        # Apply lemmatization
        lemmatized_words = [
            lemmatizer.lemmatize(word)
            for word in words
        ]

        return " ".join(lemmatized_words)


    # --------------------------------------------------------
    # 7. Apply Stemming and Lemmatization
    # --------------------------------------------------------

    for column in text_columns:

        viz_df[column + "_Stemmed"] = viz_df[column].apply(
            stemming_text
        )

        viz_df[column + "_Lemmatized"] = viz_df[column].apply(
            lemmatizing_text
        )


    # --------------------------------------------------------
    # 8. Display sample results
    # --------------------------------------------------------

    print("\n" + "=" * 90)
    print("STEMMING AND LEMMATIZATION RESULTS")
    print("=" * 90)

    for column in text_columns:

        print(f"\nOriginal Column : {column}")
        print(f"Stemmed Column  : {column}_Stemmed")
        print(f"Lemmatized Col. : {column}_Lemmatized")

        display(
            viz_df[
                [
                    column,
                    column + "_Stemmed",
                    column + "_Lemmatized"
                ]
            ].head()
        )


    # --------------------------------------------------------
    # 9. Demonstration
    # --------------------------------------------------------

    demo_text = "tourists visiting beautiful attractions"

    print("\n" + "=" * 90)
    print("NORMALIZATION EXAMPLE")
    print("=" * 90)

    print("Original      :", demo_text)
    print("Stemmed       :", stemming_text(demo_text))
    print("Lemmatized    :", lemmatizing_text(demo_text))


    # --------------------------------------------------------
    # 10. Final message
    # --------------------------------------------------------

    print("\n" + "=" * 90)
    print("✅ STEMMING AND LEMMATIZATION COMPLETED SUCCESSFULLY!")
    print("=" * 90)
````

. This can reduce vocabulary variation and make text analysis more consistent. However, in our actual Tourism dataset, no text columns were available    at the time of applying these techniques, so the techniques were demonstrated on sample text rather than applied to the final structured dataset.

````python
# ============================================================
# STEP 33: FEATURE ENGINEERING & CORRELATION REDUCTION
# Tourism Experience Analytics Project
# ============================================================

import pandas as pd
import numpy as np

print("=" * 90)
print("FEATURE ENGINEERING & CORRELATION REDUCTION")
print("=" * 90)


# ------------------------------------------------------------
# 1. Create working copy
# ------------------------------------------------------------

feature_df = outlier_df.copy()


# ------------------------------------------------------------
# 2. Convert useful columns to numeric
# ------------------------------------------------------------

numeric_candidates = [
    "VisitYear",
    "VisitMonth",
    "VisitMode",
    "AttractionId",
    "Rating"
]

for column in numeric_candidates:
    if column in feature_df.columns:
        feature_df[column] = pd.to_numeric(
            feature_df[column],
            errors="coerce"
        )


# ============================================================
# 3. CREATE NEW FEATURES
# ============================================================

# ------------------------------------------------------------
# Feature 1: Visit Quarter
# ------------------------------------------------------------

if "VisitMonth" in feature_df.columns:

    feature_df["VisitQuarter"] = (
        (feature_df["VisitMonth"] - 1) // 3 + 1
    )

    print("✅ Created: VisitQuarter")


# ------------------------------------------------------------
# Feature 2: Season
# ------------------------------------------------------------

if "VisitMonth" in feature_df.columns:

    def get_season(month):
        if pd.isna(month):
            return "Unknown"
        elif month in [12, 1, 2]:
            return "Winter"
        elif month in [3, 4, 5]:
            return "Spring"
        elif month in [6, 7, 8]:
            return "Summer"
        elif month in [9, 10, 11]:
            return "Autumn"
        else:
            return "Unknown"

    feature_df["Season"] = feature_df["VisitMonth"].apply(
        get_season
    )

    print("✅ Created: Season")


# ------------------------------------------------------------
# Feature 3: Visit Period
# ------------------------------------------------------------

if "VisitMonth" in feature_df.columns:

    def get_visit_period(month):
        if pd.isna(month):
            return "Unknown"
        elif month in [12, 1, 2, 3]:
            return "Low_Season"
        elif month in [4, 5, 6, 7, 8]:
            return "Peak_Season"
        elif month in [9, 10, 11]:
            return "Shoulder_Season"
        else:
            return "Unknown"

    feature_df["VisitPeriod"] = feature_df["VisitMonth"].apply(
        get_visit_period
    )

    print("✅ Created: VisitPeriod")


# ------------------------------------------------------------
# Feature 4: User Average Rating
# ------------------------------------------------------------

if "UserId" in feature_df.columns and "Rating" in feature_df.columns:

    user_avg_rating = (
        feature_df.groupby("UserId")["Rating"]
        .transform("mean")
    )

    feature_df["UserAvgRating"] = user_avg_rating

    print("✅ Created: UserAvgRating")


# ------------------------------------------------------------
# Feature 5: User Visit Count
# ------------------------------------------------------------

if "UserId" in feature_df.columns:

    user_visit_count = (
        feature_df.groupby("UserId")["UserId"]
        .transform("count")
    )

    feature_df["UserVisitCount"] = user_visit_count

    print("✅ Created: UserVisitCount")


# ------------------------------------------------------------
# Feature 6: Attraction Average Rating
# ------------------------------------------------------------

if (
    "AttractionId" in feature_df.columns
    and "Rating" in feature_df.columns
):

    attraction_avg_rating = (
        feature_df.groupby("AttractionId")["Rating"]
        .transform("mean")
    )

    feature_df["AttractionAvgRating"] = (
        attraction_avg_rating
    )

    print("✅ Created: AttractionAvgRating")


# ------------------------------------------------------------
# Feature 7: Attraction Visit Count
# ------------------------------------------------------------

if "AttractionId" in feature_df.columns:

    attraction_visit_count = (
        feature_df.groupby("AttractionId")["AttractionId"]
        .transform("count")
    )

    feature_df["AttractionVisitCount"] = (
        attraction_visit_count
    )

    print("✅ Created: AttractionVisitCount")


# ------------------------------------------------------------
# Feature 8: Rating Difference from User Average
# ------------------------------------------------------------

if (
    "Rating" in feature_df.columns
    and "UserAvgRating" in feature_df.columns
):

    feature_df["RatingDifference"] = (
        feature_df["Rating"]
        - feature_df["UserAvgRating"]
    )

    print("✅ Created: RatingDifference")


# ============================================================
# 4. CHECK NUMERICAL CORRELATION
# ============================================================

print("\n" + "=" * 90)
print("NUMERICAL CORRELATION ANALYSIS")
print("=" * 90)

numeric_features = feature_df.select_dtypes(
    include=np.number
)

correlation_matrix = numeric_features.corr()

display(correlation_matrix.round(2))


# ============================================================
# 5. FIND HIGHLY CORRELATED FEATURES
# ============================================================

correlation_threshold = 0.90

upper_triangle = correlation_matrix.where(
    np.triu(
        np.ones(correlation_matrix.shape),
        k=1
    ).astype(bool)
)

high_correlation_pairs = []

for column in upper_triangle.columns:

    for index in upper_triangle.index:

        correlation_value = upper_triangle.loc[index, column]

        if pd.notna(correlation_value):
            if abs(correlation_value) >= correlation_threshold:

                high_correlation_pairs.append({
                    "Feature 1": index,
                    "Feature 2": column,
                    "Correlation": round(
                        correlation_value, 3
                    )
                })

high_corr_df = pd.DataFrame(
    high_correlation_pairs
)

print("\nHighly Correlated Feature Pairs:")
print(f"Threshold: |correlation| >= {correlation_threshold}")

if high_corr_df.empty:
    print("✅ No highly correlated feature pairs found.")
else:
    display(high_corr_df)


# ============================================================
# 6. REMOVE REDUNDANT HIGHLY CORRELATED FEATURES
# ============================================================

features_to_remove = set()

if not high_corr_df.empty:

    for _, row in high_corr_df.iterrows():

        feature1 = row["Feature 1"]
        feature2 = row["Feature 2"]

        # Keep the first feature and remove the second
        # only when both are available and not essential
        if feature2 not in [
            "Rating",
            "UserAvgRating",
            "AttractionAvgRating"
        ]:
            features_to_remove.add(feature2)


# Remove redundant features
if features_to_remove:

    feature_df = feature_df.drop(
        columns=list(features_to_remove),
        errors="ignore"
    )

    print(
        "\n✅ Removed redundant features:",
        list(features_to_remove)
    )

else:
    print("\n✅ No redundant features removed.")


# ============================================================
# 7. FINAL FEATURE LIST
# ============================================================

print("\n" + "=" * 90)
print("FINAL FEATURE SET")
print("=" * 90)

print("Number of rows    :", feature_df.shape[0])
print("Number of columns :", feature_df.shape[1])

print("\nFeatures:")
print(feature_df.columns.tolist())


# ============================================================
# 8. DISPLAY SAMPLE DATA
# ============================================================

print("\n" + "=" * 90)
print("FEATURE ENGINEERING SAMPLE")
print("=" * 90)

display(feature_df.head())


# ============================================================
# 9. SAVE FEATURE-ENGINEERED DATASET
# ============================================================

feature_df.to_csv(
    data_path + r"\tourism_feature_engineered.csv",
    index=False
)

print("\n✅ Feature-engineered dataset saved successfully:")
print(data_path + r"\tourism_feature_engineered.csv")


print("\n" + "=" * 90)
print("✅ FEATURE ENGINEERING & CORRELATION REDUCTION COMPLETED!")
print("=" * 90)# Manipulate Features to minimize feature correlation and create new features
````

````python
# ============================================================
# STEP 38: DIMENSIONALITY REDUCTION USING PCA
# Tourism Experience Analytics Project
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

print("=" * 90)
print("DIMENSIONALITY REDUCTION USING PCA")
print("=" * 90)


# ------------------------------------------------------------
# 1. Check whether feature_df exists
# ------------------------------------------------------------

if "feature_df" not in globals():

    print("⚠️ feature_df is not defined.")
    print("Please run the Feature Engineering step first.")

else:

    # --------------------------------------------------------
    # 2. Create working copy
    # --------------------------------------------------------

    pca_df = feature_df.copy()


    # --------------------------------------------------------
    # 3. Select numerical features
    # --------------------------------------------------------

    numeric_columns = pca_df.select_dtypes(
        include=np.number
    ).columns.tolist()

    # Remove target variable
    if "Rating" in numeric_columns:
        numeric_columns.remove("Rating")

    # Remove identifier columns
    id_columns = [
        "TransactionId",
        "UserId",
        "AttractionId",
        "CityId",
        "CountryId",
        "RegionId",
        "ContinentId",
        "VisitModeId",
        "AttractionTypeId"
    ]

    numeric_columns = [
        col for col in numeric_columns
        if col not in id_columns
    ]

    print("\nNumerical features selected:")
    print(numeric_columns)


    # --------------------------------------------------------
    # 4. Check whether enough features exist
    # --------------------------------------------------------

    if len(numeric_columns) < 2:

        print(
            "\n⚠️ PCA is not required because fewer than "
            "two suitable numerical features are available."
        )

    else:

        # ----------------------------------------------------
        # 5. Prepare data
        # ----------------------------------------------------

        X = pca_df[numeric_columns].copy()

        # Replace infinite values
        X = X.replace(
            [np.inf, -np.inf],
            np.nan
        )

        # Fill missing values with median
        X = X.fillna(
            X.median(numeric_only=True)
        )


        # ----------------------------------------------------
        # 6. Standardize before PCA
        # ----------------------------------------------------

        scaler_pca = StandardScaler()

        X_scaled = scaler_pca.fit_transform(X)


        # ----------------------------------------------------
        # 7. Fit PCA
        # ----------------------------------------------------

        # Keep enough components to explain 95% variance
        pca = PCA(
            n_components=0.95,
            random_state=42
        )

        X_pca = pca.fit_transform(X_scaled)


        # ----------------------------------------------------
        # 8. Create PCA dataframe
        # ----------------------------------------------------

        pca_columns = [
            f"PC{i + 1}"
            for i in range(X_pca.shape[1])
        ]

        pca_result = pd.DataFrame(
            X_pca,
            columns=pca_columns,
            index=pca_df.index
        )


        # ----------------------------------------------------
        # 9. Display PCA information
        # ----------------------------------------------------

        print("\n" + "=" * 90)
        print("PCA RESULTS")
        print("=" * 90)

        print(
            "Original number of features :",
            len(numeric_columns)
        )

        print(
            "Reduced number of features  :",
            X_pca.shape[1]
        )

        print(
            "Explained variance ratio:"
        )

        for i, variance in enumerate(
            pca.explained_variance_ratio_,
            start=1
        ):
            print(
                f"PC{i}: {variance:.4f} "
                f"({variance * 100:.2f}%)"
            )

        print(
            "\nTotal explained variance:",
            f"{pca.explained_variance_ratio_.sum() * 100:.2f}%"
        )


        # ----------------------------------------------------
        # 10. Display transformed data
        # ----------------------------------------------------

        print("\nPCA Transformed Data:")
        display(pca_result.head())


        # ----------------------------------------------------
        # 11. Plot explained variance
        # ----------------------------------------------------

        cumulative_variance = np.cumsum(
            pca.explained_variance_ratio_
        )

        plt.figure(figsize=(10, 5))

        plt.plot(
            range(1, len(cumulative_variance) + 1),
            cumulative_variance,
            marker="o"
        )

        plt.axhline(
            y=0.95,
            linestyle="--"
        )

        plt.title(
            "Cumulative Explained Variance using PCA"
        )

        plt.xlabel(
            "Number of Principal Components"
        )

        plt.ylabel(
            "Cumulative Explained Variance"
        )

        plt.xticks(
            range(1, len(cumulative_variance) + 1)
        )

        plt.grid(True)

        plt.tight_layout()
        plt.show()


        # ----------------------------------------------------
        # 12. Display feature contributions
        # ----------------------------------------------------

        print("\n" + "=" * 90)
        print("PCA FEATURE CONTRIBUTIONS")
        print("=" * 90)

        components_df = pd.DataFrame(
            pca.components_,
            columns=numeric_columns,
            index=pca_columns
        )

        display(
            components_df.round(4)
        )


        # ----------------------------------------------------
        # 13. Save PCA results
        # ----------------------------------------------------

        pca_file = (
            data_path +
            r"\tourism_pca_data.csv"
        )

        pca_result.to_csv(
            pca_file,
            index=False
        )

        print("\n✅ PCA dataset saved to:")
        print(pca_file)


        # Save PCA and scaler
        import joblib

        pca_model_file = (
            data_path +
            r"\tourism_pca_model.pkl"
        )

        pca_scaler_file = (
            data_path +
            r"\tourism_pca_scaler.pkl"
        )

        joblib.dump(
            pca,
            pca_model_file
        )

        joblib.dump(
            scaler_pca,
            pca_scaler_file
        )

        print("\n✅ PCA model saved to:")
        print(pca_model_file)

        print("\n✅ PCA scaler saved to:")
        print(pca_scaler_file)


        print("\n" + "=" * 90)
        print("✅ DIMENSIONALITY REDUCTION COMPLETED SUCCESSFULLY!")
        print("=" * 90)# DImensionality Reduction (If needed)
````

### 9. Handling Imbalanced Dataset

##### Do you think the dataset is imbalanced? Explain Why.

. We checked the distribution of the VisitMode classes using class counts and percentages. The dataset is considered imbalanced only when there is a     substantial difference between the majority and minority classes.
. This is important because an imbalanced classification dataset can cause the model to favor the majority class and perform poorly on minority . .classes. Therefore, we checked class distribution before deciding whether balancing was required.

````python
# ============================================================
# STEP 40: HANDLING IMBALANCED DATASET
# Tourism Experience Analytics Project
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from imblearn.over_sampling import RandomOverSampler

print("=" * 90)
print("IMBALANCED DATASET CHECK & HANDLING")
print("=" * 90)


# ------------------------------------------------------------
# 1. Load Transaction Dataset
# ------------------------------------------------------------

data_path = r"C:\Users\Admin\OneDrive\Desktop\TOURISM_PROJECT\DATA"

transaction_df = pd.read_csv(
    data_path + r"\Transaction.csv",
    encoding="latin1"
)

print("\nDataset Shape:")
print(transaction_df.shape)


# ------------------------------------------------------------
# 2. Check VisitMode column
# ------------------------------------------------------------

if "VisitMode" not in transaction_df.columns:

    print("❌ VisitMode column not found.")
    
else:

    # --------------------------------------------------------
    # 3. Remove missing target values
    # --------------------------------------------------------

    classification_df = transaction_df.dropna(
        subset=["VisitMode"]
    ).copy()


    # --------------------------------------------------------
    # 4. Display original class distribution
    # --------------------------------------------------------

    class_counts = classification_df["VisitMode"].value_counts()

    print("\n" + "=" * 90)
    print("ORIGINAL VISIT MODE DISTRIBUTION")
    print("=" * 90)

    print(class_counts)


    # --------------------------------------------------------
    # 5. Calculate class percentages
    # --------------------------------------------------------

    class_percentage = (
        classification_df["VisitMode"]
        .value_counts(normalize=True) * 100
    )

    print("\nClass Percentage:")
    print(class_percentage.round(2))


    # --------------------------------------------------------
    # 6. Calculate imbalance ratio
    # --------------------------------------------------------

    majority_count = class_counts.max()
    minority_count = class_counts.min()

    imbalance_ratio = (
        majority_count / minority_count
    )

    print("\n" + "=" * 90)
    print("IMBALANCE ANALYSIS")
    print("=" * 90)

    print("Majority Class Count :", majority_count)
    print("Minority Class Count :", minority_count)
    print(f"Imbalance Ratio      : {imbalance_ratio:.2f}")


    # --------------------------------------------------------
    # 7. Visualize original distribution
    # --------------------------------------------------------

    plt.figure(figsize=(10, 5))

    plt.bar(
        class_counts.index.astype(str),
        class_counts.values
    )

    plt.title(
        "Visit Mode Distribution Before Balancing"
    )

    plt.xlabel("Visit Mode")
    plt.ylabel("Number of Records")

    plt.xticks(rotation=45)

    plt.tight_layout()
    plt.show()


    # --------------------------------------------------------
    # 8. Create features and target
    # --------------------------------------------------------

    X = classification_df.drop(
        columns=["VisitMode"],
        errors="ignore"
    )

    y = classification_df["VisitMode"]


    # --------------------------------------------------------
    # 9. Remove identifier columns
    # --------------------------------------------------------

    id_columns = [
        "TransactionId",
        "UserId",
        "AttractionId"
    ]

    X = X.drop(
        columns=[
            col for col in id_columns
            if col in X.columns
        ],
        errors="ignore"
    )


    # --------------------------------------------------------
    # 10. Keep only numerical features for demonstration
    # --------------------------------------------------------

    X_numeric = X.select_dtypes(
        include=np.number
    ).copy()


    # --------------------------------------------------------
    # 11. Handle missing numerical values
    # --------------------------------------------------------

    X_numeric = X_numeric.replace(
        [np.inf, -np.inf],
        np.nan
    )

    X_numeric = X_numeric.fillna(
        X_numeric.median(numeric_only=True)
    )


    # --------------------------------------------------------
    # 12. Train-test split BEFORE oversampling
    # --------------------------------------------------------

    X_train, X_test, y_train, y_test = train_test_split(
        X_numeric,
        y,
        test_size=0.20,
        random_state=42,
        stratify=y
    )


    # --------------------------------------------------------
    # 13. Check training distribution
    # --------------------------------------------------------

    print("\n" + "=" * 90)
    print("TRAINING DISTRIBUTION BEFORE BALANCING")
    print("=" * 90)

    print(y_train.value_counts())


    # --------------------------------------------------------
    # 14. Decide whether balancing is needed
    # --------------------------------------------------------

    train_counts = y_train.value_counts()

    train_ratio = (
        train_counts.max() /
        train_counts.min()
    )

    print(
        f"\nTraining Imbalance Ratio: {train_ratio:.2f}"
    )


    # --------------------------------------------------------
    # 15. Apply Random Oversampling if needed
    # --------------------------------------------------------

    if train_ratio > 1.50:

        print("\n⚠️ Dataset is imbalanced.")
        print("Applying Random Oversampling...")

        ros = RandomOverSampler(
            random_state=42
        )

        X_train_balanced, y_train_balanced = ros.fit_resample(
            X_train,
            y_train
        )

        print("\n✅ Oversampling completed.")

    else:

        print("\n✅ Dataset is reasonably balanced.")
        print("No oversampling is required.")

        X_train_balanced = X_train.copy()
        y_train_balanced = y_train.copy()


    # --------------------------------------------------------
    # 16. Display distribution after balancing
    # --------------------------------------------------------

    print("\n" + "=" * 90)
    print("TRAINING DISTRIBUTION AFTER BALANCING")
    print("=" * 90)

    print(
        y_train_balanced.value_counts()
    )


    # --------------------------------------------------------
    # 17. Visualize balanced distribution
    # --------------------------------------------------------

    balanced_counts = (
        y_train_balanced.value_counts()
    )

    plt.figure(figsize=(10, 5))

    plt.bar(
        balanced_counts.index.astype(str),
        balanced_counts.values
    )

    plt.title(
        "Visit Mode Distribution After Balancing"
    )

    plt.xlabel("Visit Mode")
    plt.ylabel("Number of Training Records")

    plt.xticks(rotation=45)

    plt.tight_layout()
    plt.show()


    # --------------------------------------------------------
    # 18. Final result
    # --------------------------------------------------------

    print("\n" + "=" * 90)
    print("FINAL RESULT")
    print("=" * 90)

    print(
        "Original training records:",
        len(y_train)
    )

    print(
        "Balanced training records:",
        len(y_train_balanced)
    )

    print(
        "Test records:",
        len(y_test)
    )

    print("\n✅ IMBALANCED DATASET CHECK/HANDLING COMPLETED!")
    print("=" * 90)
````

````python
# ============================================================
# ML MODEL - 1 IMPLEMENTATION
# RANDOM FOREST REGRESSION
# Target: Rating
# ============================================================

import os
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

print("=" * 90)
print("ML MODEL - 1: RANDOM FOREST REGRESSION")
print("=" * 90)


# ------------------------------------------------------------
# 1. Dataset Path
# ------------------------------------------------------------

data_path = r"C:\Users\Admin\OneDrive\Desktop\TOURISM_PROJECT\DATA"


# ------------------------------------------------------------
# 2. Load Required Datasets
# ------------------------------------------------------------

transaction_df = pd.read_csv(
    os.path.join(data_path, "Transaction.csv"),
    encoding="latin1"
)

user_df = pd.read_csv(
    os.path.join(data_path, "User.csv"),
    encoding="latin1"
)

item_df = pd.read_csv(
    os.path.join(data_path, "Item.csv"),
    encoding="latin1"
)

type_df = pd.read_csv(
    os.path.join(data_path, "Type.csv"),
    encoding="latin1"
)

mode_df = pd.read_csv(
    os.path.join(data_path, "Mode.csv"),
    encoding="latin1"
)


# ------------------------------------------------------------
# 3. Create Consolidated Dataset
# ------------------------------------------------------------

model_df = transaction_df.copy()

# Merge User data
model_df = model_df.merge(
    user_df,
    on="UserId",
    how="left"
)

# Merge Attraction data
item_clean = item_df.drop_duplicates(
    subset=["AttractionId"]
)

model_df = model_df.merge(
    item_clean[
        [
            "AttractionId",
            "AttractionCityId",
            "AttractionTypeId",
            "Attraction"
        ]
    ],
    on="AttractionId",
    how="left"
)

# Merge Attraction Type
type_clean = type_df.drop_duplicates(
    subset=["AttractionTypeId"]
)

model_df = model_df.merge(
    type_clean,
    on="AttractionTypeId",
    how="left"
)

# Merge Visit Mode
mode_clean = mode_df.drop_duplicates(
    subset=["VisitModeId"]
)

model_df = model_df.merge(
    mode_clean[
        ["VisitModeId", "VisitMode"]
    ],
    left_on="VisitMode",
    right_on="VisitModeId",
    how="left",
    suffixes=("_id", "_name")
)

print("\nConsolidated dataset shape:")
print(model_df.shape)


# ============================================================
# 4. SELECT FEATURES AND TARGET
# ============================================================

target = "Rating"

feature_columns = [
    "VisitYear",
    "VisitMonth",
    "VisitMode",
    "ContinentId",
    "RegionId",
    "CountryId",
    "CityId",
    "AttractionCityId",
    "AttractionTypeId"
]

# Keep only columns that exist
feature_columns = [
    col
    for col in feature_columns
    if col in model_df.columns
]

# Create modeling dataframe
model_data = model_df[
    feature_columns + [target]
].copy()

# Convert target to numeric
model_data[target] = pd.to_numeric(
    model_data[target],
    errors="coerce"
)

# Remove rows with missing target
model_data = model_data.dropna(
    subset=[target]
)

print("\nFeatures used:")
print(feature_columns)

print("\nTarget:")
print(target)


# ============================================================
# 5. SPLIT X AND Y
# ============================================================

X = model_data[feature_columns].copy()
y = model_data[target].copy()


# ============================================================
# 6. IDENTIFY NUMERICAL & CATEGORICAL FEATURES
# ============================================================

numeric_features = [
    col
    for col in feature_columns
    if pd.api.types.is_numeric_dtype(X[col])
]

categorical_features = [
    col
    for col in feature_columns
    if col not in numeric_features
]

print("\nNumerical features:")
print(numeric_features)

print("\nCategorical features:")
print(categorical_features)


# ============================================================
# 7. PREPROCESSING
# ============================================================

numeric_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="median")
    )
])

categorical_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="most_frequent")
    ),
    (
        "encoder",
        OneHotEncoder(
            handle_unknown="ignore"
        )
    )
])

preprocessor = ColumnTransformer([
    (
        "numeric",
        numeric_pipeline,
        numeric_features
    ),
    (
        "categorical",
        categorical_pipeline,
        categorical_features
    )
])


# ============================================================
# 8. CREATE RANDOM FOREST REGRESSOR
# ============================================================

rf_regressor = RandomForestRegressor(
    n_estimators=150,
    max_depth=20,
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1
)


# ============================================================
# 9. CREATE COMPLETE PIPELINE
# ============================================================

rf_model = Pipeline([
    (
        "preprocessor",
        preprocessor
    ),
    (
        "regressor",
        rf_regressor
    )
])


# ============================================================
# 10. SPLIT DATA INTO TRAIN AND TEST
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\n" + "=" * 90)
print("TRAIN-TEST SPLIT")
print("=" * 90)

print("Training rows:", len(X_train))
print("Testing rows :", len(X_test))


# ============================================================
# 11. FIT THE ALGORITHM
# ============================================================

print("\n" + "=" * 90)
print("FITTING RANDOM FOREST REGRESSION MODEL")
print("=" * 90)

rf_model.fit(
    X_train,
    y_train
)

print("✅ Random Forest Regressor fitted successfully!")


# ============================================================
# 12. PREDICT ON THE MODEL
# ============================================================

print("\n" + "=" * 90)
print("PREDICTION ON TEST DATA")
print("=" * 90)

y_pred = rf_model.predict(
    X_test
)

print("✅ Predictions generated successfully!")


# ============================================================
# 13. DISPLAY ACTUAL VS PREDICTED VALUES
# ============================================================

prediction_result = pd.DataFrame({
    "Actual Rating": y_test.values,
    "Predicted Rating": y_pred
})

print("\nActual vs Predicted Ratings:")
display(
    prediction_result.head(20)
)


# ============================================================
# 14. MODEL EVALUATION
# ============================================================

mae = mean_absolute_error(
    y_test,
    y_pred
)

mse = mean_squared_error(
    y_test,
    y_pred
)

rmse = np.sqrt(mse)

r2 = r2_score(
    y_test,
    y_pred
)

print("\n" + "=" * 90)
print("MODEL EVALUATION")
print("=" * 90)

print(f"MAE  : {mae:.4f}")
print(f"MSE  : {mse:.4f}")
print(f"RMSE : {rmse:.4f}")
print(f"R²   : {r2:.4f}")


# ============================================================
# 15. SAMPLE PREDICTION
# ============================================================

print("\n" + "=" * 90)
print("SAMPLE PREDICTION")
print("=" * 90)

sample_data = X_test.head(5)

sample_predictions = rf_model.predict(
    sample_data
)

sample_output = sample_data.copy()

sample_output["Predicted Rating"] = (
    sample_predictions
)

display(
    sample_output
)


# ============================================================
# 16. SAVE MODEL
# ============================================================

import joblib

model_file = os.path.join(
    data_path,
    "tourism_rating_random_forest_model.pkl"
)

joblib.dump(
    rf_model,
    model_file
)

print("\n✅ Model saved successfully:")
print(model_file)


print("\n" + "=" * 90)
print("✅ ML MODEL - 1 COMPLETED SUCCESSFULLY!")
print("=" * 90)# ML Model - 1 Implementation

# Fit the Algorithm

# Predict on the model
````

#### 2. Cross- Validation & Hyperparameter Tuning

````python
# ============================================================
# ML MODEL - 2
# CROSS-VALIDATION & HYPERPARAMETER TUNING
# Random Forest Classifier - VisitMode Prediction
# ============================================================

import os
import joblib
import pandas as pd
import numpy as np

from sklearn.model_selection import (
    GridSearchCV,
    StratifiedKFold,
    train_test_split
)

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer

from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

print("=" * 90)
print("MODEL 2 - CROSS-VALIDATION & HYPERPARAMETER TUNING")
print("=" * 90)


# ============================================================
# 1. PATHS
# ============================================================

project_path = r"C:\Users\Admin\OneDrive\Desktop\TOURISM_PROJECT"
data_path = os.path.join(project_path, "DATA")
model_path = os.path.join(project_path, "models")

os.makedirs(model_path, exist_ok=True)


# ============================================================
# 2. LOAD DATA
# ============================================================

transaction_df = pd.read_csv(
    os.path.join(data_path, "Transaction.csv"),
    encoding="latin1"
)

user_df = pd.read_csv(
    os.path.join(data_path, "User.csv"),
    encoding="latin1"
)

item_df = pd.read_csv(
    os.path.join(data_path, "Item.csv"),
    encoding="latin1"
)

type_df = pd.read_csv(
    os.path.join(data_path, "Type.csv"),
    encoding="latin1"
)

mode_df = pd.read_csv(
    os.path.join(data_path, "Mode.csv"),
    encoding="latin1"
)


# ============================================================
# 3. MERGE DATASETS
# ============================================================

df = transaction_df.copy()

# User information
df = df.merge(
    user_df,
    on="UserId",
    how="left"
)

# Attraction information
item_clean = (
    item_df[
        [
            "AttractionId",
            "AttractionCityId",
            "AttractionTypeId"
        ]
    ]
    .drop_duplicates("AttractionId")
)

df = df.merge(
    item_clean,
    on="AttractionId",
    how="left"
)

# Attraction type
type_clean = (
    type_df[
        [
            "AttractionTypeId",
            "AttractionType"
        ]
    ]
    .drop_duplicates("AttractionTypeId")
)

df = df.merge(
    type_clean,
    on="AttractionTypeId",
    how="left"
)

# Visit mode
mode_clean = (
    mode_df[
        [
            "VisitModeId",
            "VisitMode"
        ]
    ]
    .drop_duplicates("VisitModeId")
)

df = df.merge(
    mode_clean,
    left_on="VisitMode",
    right_on="VisitModeId",
    how="left",
    suffixes=("_id", "_name")
)


# ============================================================
# 4. TARGET
# ============================================================

target = "VisitMode_name"

if target not in df.columns:
    raise ValueError(
        f"{target} was not created. Check Mode.csv."
    )

df = df.dropna(
    subset=[target]
)

y = df[target].astype(str)


# ============================================================
# 5. SELECT FEATURES
# ============================================================

features = [
    "VisitYear",
    "VisitMonth",
    "ContinentId",
    "RegionId",
    "CountryId",
    "CityId",
    "AttractionCityId",
    "AttractionTypeId"
]

features = [
    col for col in features
    if col in df.columns
]

X = df[features].copy()


print("\nFeatures:")
print(features)

print("\nTarget:")
print(target)


# ============================================================
# 6. TRAIN / TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining records:", len(X_train))
print("Testing records :", len(X_test))


# ============================================================
# 7. IDENTIFY FEATURE TYPES
# ============================================================

numeric_features = X_train.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object", "string", "category"]
).columns.tolist()

print("\nNumeric features:")
print(numeric_features)

print("\nCategorical features:")
print(categorical_features)


# ============================================================
# 8. PREPROCESSING
# ============================================================

numeric_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="median")
    )
])

categorical_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="most_frequent")
    ),
    (
        "encoder",
        OneHotEncoder(
            handle_unknown="ignore"
        )
    )
])

preprocessor = ColumnTransformer([
    (
        "numeric",
        numeric_pipeline,
        numeric_features
    ),
    (
        "categorical",
        categorical_pipeline,
        categorical_features
    )
])


# ============================================================
# 9. RANDOM FOREST CLASSIFIER
# ============================================================

rf_classifier = RandomForestClassifier(
    random_state=42,
    n_jobs=-1,
    class_weight="balanced"
)


# ============================================================
# 10. CREATE PIPELINE
# ============================================================

classification_pipeline = Pipeline([
    (
        "preprocessor",
        preprocessor
    ),
    (
        "classifier",
        rf_classifier
    )
])


# ============================================================
# 11. STRATIFIED 5-FOLD CROSS-VALIDATION
# ============================================================

skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)


# ============================================================
# 12. HYPERPARAMETER GRID
# ============================================================

param_grid = {
    "classifier__n_estimators": [
        50,
        100
    ],

    "classifier__max_depth": [
        10,
        20
    ],

    "classifier__min_samples_split": [
        2,
        5
    ],

    "classifier__min_samples_leaf": [
        1,
        2
    ]
}

print("\n" + "=" * 90)
print("HYPERPARAMETERS")
print("=" * 90)

for parameter, values in param_grid.items():
    print(parameter, ":", values)


# ============================================================
# 13. GRID SEARCH
# ============================================================

grid_search = GridSearchCV(
    estimator=classification_pipeline,
    param_grid=param_grid,
    scoring="f1_weighted",
    cv=skf,
    n_jobs=-1,
    verbose=1,
    return_train_score=True
)


# ============================================================
# 14. FIT GRID SEARCH
# ============================================================

print("\n" + "=" * 90)
print("STARTING GRID SEARCH")
print("=" * 90)

grid_search.fit(
    X_train,
    y_train
)

print("\n✅ Grid Search completed successfully!")


# ============================================================
# 15. BEST PARAMETERS
# ============================================================

print("\n" + "=" * 90)
print("BEST PARAMETERS")
print("=" * 90)

print(
    grid_search.best_params_
)


# ============================================================
# 16. BEST CV SCORE
# ============================================================

best_cv_f1 = grid_search.best_score_

print(
    "\nBest 5-Fold CV Weighted F1:",
    round(best_cv_f1, 4)
)


# ============================================================
# 17. BEST MODEL
# ============================================================

best_classification_model = (
    grid_search.best_estimator_
)


# ============================================================
# 18. TEST PREDICTION
# ============================================================

y_pred = best_classification_model.predict(
    X_test
)


# ============================================================
# 19. EVALUATION
# ============================================================

accuracy = accuracy_score(
    y_test,
    y_pred
)

precision = precision_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)


print("\n" + "=" * 90)
print("TUNED MODEL TEST PERFORMANCE")
print("=" * 90)

print(f"Accuracy  : {accuracy:.4f}")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1-Score  : {f1:.4f}")


# ============================================================
# 20. CLASSIFICATION REPORT
# ============================================================

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred,
        zero_division=0
    )
)


# ============================================================
# 21. GRID SEARCH RESULTS
# ============================================================

results_df = pd.DataFrame(
    grid_search.cv_results_
)

results_df = results_df[
    [
        "params",
        "mean_test_score",
        "std_test_score",
        "rank_test_score"
    ]
].copy()

results_df["Mean CV F1"] = (
    results_df["mean_test_score"]
)

results_df["Std CV F1"] = (
    results_df["std_test_score"]
)

results_df = results_df[
    [
        "params",
        "Mean CV F1",
        "Std CV F1",
        "rank_test_score"
    ]
].sort_values(
    "Mean CV F1",
    ascending=False
)

print("\n" + "=" * 90)
print("TOP HYPERPARAMETER COMBINATIONS")
print("=" * 90)

display(
    results_df.head(10).round(4)
)


# ============================================================
# 22. SAVE TUNED MODEL
# ============================================================

tuned_model_file = os.path.join(
    model_path,
    "tourism_visitmode_tuned_model.pkl"
)

joblib.dump(
    best_classification_model,
    tuned_model_file,
    compress=3
)

print("\n✅ Tuned model saved:")
print(tuned_model_file)


# ============================================================
# 23. SAVE BEST PARAMETERS
# ============================================================

parameter_file = os.path.join(
    model_path,
    "tourism_visitmode_best_parameters.pkl"
)

joblib.dump(
    grid_search.best_params_,
    parameter_file
)

print("\n✅ Best parameters saved:")
print(parameter_file)


# ============================================================
# 24. SAVE EVALUATION RESULTS
# ============================================================

evaluation_df = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1-Score",
        "Best CV F1"
    ],

    "Score": [
        accuracy,
        precision,
        recall,
        f1,
        best_cv_f1
    ]
})

evaluation_file = os.path.join(
    model_path,
    "tourism_visitmode_tuned_evaluation.csv"
)

evaluation_df.to_csv(
    evaluation_file,
    index=False
)

print("\n✅ Evaluation results saved:")
print(evaluation_file)


# ============================================================
# 25. SAVE CLASS LABELS
# ============================================================

classes_file = os.path.join(
    model_path,
    "tourism_visitmode_classes.joblib"
)

joblib.dump(
    sorted(y.unique().tolist()),
    classes_file
)

print("\n✅ Class labels saved:")
print(classes_file)


# ============================================================
# 26. VERIFY SAVED MODEL
# ============================================================

loaded_model = joblib.load(
    tuned_model_file
)

print("\n✅ Saved tuned model loaded successfully!")


# ============================================================
# 27. SANITY CHECK
# ============================================================

sample_prediction = loaded_model.predict(
    X_test.head(1)
)

print(
    "Sample VisitMode Prediction:",
    sample_prediction[0]
)


print("\n" + "=" * 90)
print("✅ MODEL 2 CROSS-VALIDATION & HYPERPARAMETER TUNING COMPLETE!")
print("=" * 90)
````

````python
# ============================================================
# ML MODEL - 2
# CLASSIFICATION: VISIT MODE PREDICTION
# Evaluation Metrics + Cross-Validation + Score Chart
# Streamlit Cloud Preparation
# ============================================================

import os
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    cross_validate
)

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer

from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

print("=" * 100)
print("ML MODEL - 2: RANDOM FOREST CLASSIFICATION")
print("TARGET: VISIT MODE")
print("=" * 100)


# ============================================================
# 1. PATHS
# ============================================================

project_path = r"C:\Users\Admin\OneDrive\Desktop\TOURISM_PROJECT"
data_path = os.path.join(project_path, "DATA")
model_path = os.path.join(project_path, "models")

os.makedirs(model_path, exist_ok=True)


# ============================================================
# 2. LOAD DATA
# ============================================================

transaction_df = pd.read_csv(
    os.path.join(data_path, "Transaction.csv"),
    encoding="latin1"
)

user_df = pd.read_csv(
    os.path.join(data_path, "User.csv"),
    encoding="latin1"
)

item_df = pd.read_csv(
    os.path.join(data_path, "Item.csv"),
    encoding="latin1"
)

type_df = pd.read_csv(
    os.path.join(data_path, "Type.csv"),
    encoding="latin1"
)

mode_df = pd.read_csv(
    os.path.join(data_path, "Mode.csv"),
    encoding="latin1"
)


# ============================================================
# 3. CREATE CONSOLIDATED DATASET
# ============================================================

cls_df = transaction_df.copy()

# User information
cls_df = cls_df.merge(
    user_df,
    on="UserId",
    how="left"
)

# Attraction information
item_clean = item_df.drop_duplicates(
    subset=["AttractionId"]
)

cls_df = cls_df.merge(
    item_clean[
        [
            "AttractionId",
            "AttractionCityId",
            "AttractionTypeId"
        ]
    ],
    on="AttractionId",
    how="left"
)

# Attraction type
type_clean = type_df.drop_duplicates(
    subset=["AttractionTypeId"]
)

cls_df = cls_df.merge(
    type_clean,
    on="AttractionTypeId",
    how="left"
)

# Visit mode name
mode_clean = mode_df.drop_duplicates(
    subset=["VisitModeId"]
)

cls_df = cls_df.merge(
    mode_clean[
        ["VisitModeId", "VisitMode"]
    ],
    left_on="VisitMode",
    right_on="VisitModeId",
    how="left",
    suffixes=("_id", "_name")
)

print("\nConsolidated Dataset Shape:")
print(cls_df.shape)


# ============================================================
# 4. TARGET COLUMN
# ============================================================

if "VisitMode_name" in cls_df.columns:
    target = "VisitMode_name"
else:
    target = "VisitMode"

print("\nTarget Variable:")
print(target)


# ============================================================
# 5. SELECT FEATURES
# ============================================================

features = [
    "VisitYear",
    "VisitMonth",
    "ContinentId",
    "RegionId",
    "CountryId",
    "CityId",
    "AttractionCityId",
    "AttractionTypeId"
]

features = [
    col
    for col in features
    if col in cls_df.columns
]

model_data = cls_df[
    features + [target]
].copy()

model_data = model_data.dropna(
    subset=[target]
)

X = model_data[features].copy()
y = model_data[target].copy()


# ============================================================
# 6. REMOVE INVALID TARGET STRINGS
# ============================================================

y = y.astype(str)

valid_rows = (
    y.notna()
    & (y.str.strip() != "")
    & (y.str.lower() != "nan")
)

X = X.loc[valid_rows].copy()
y = y.loc[valid_rows].copy()


# ============================================================
# 7. IDENTIFY FEATURE TYPES
# ============================================================

numeric_features = [
    col
    for col in features
    if pd.api.types.is_numeric_dtype(X[col])
]

categorical_features = [
    col
    for col in features
    if col not in numeric_features
]

print("\nNumerical Features:")
print(numeric_features)

print("\nCategorical Features:")
print(categorical_features)


# ============================================================
# 8. PREPROCESSING PIPELINE
# ============================================================

numeric_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="median")
    )
])

categorical_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="most_frequent")
    ),
    (
        "encoder",
        OneHotEncoder(
            handle_unknown="ignore"
        )
    )
])

preprocessor = ColumnTransformer([
    (
        "numeric",
        numeric_pipeline,
        numeric_features
    ),
    (
        "categorical",
        categorical_pipeline,
        categorical_features
    )
])


# ============================================================
# 9. RANDOM FOREST CLASSIFIER
# ============================================================

classifier = RandomForestClassifier(
    n_estimators=150,
    max_depth=20,
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1,
    class_weight="balanced"
)

classification_model = Pipeline([
    (
        "preprocessor",
        preprocessor
    ),
    (
        "classifier",
        classifier
    )
])


# ============================================================
# 10. TRAIN-TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\n" + "=" * 100)
print("TRAIN-TEST SPLIT")
print("=" * 100)

print("Training Records :", len(X_train))
print("Testing Records  :", len(X_test))


# ============================================================
# 11. FIT MODEL
# ============================================================

print("\n" + "=" * 100)
print("FITTING RANDOM FOREST CLASSIFIER")
print("=" * 100)

classification_model.fit(
    X_train,
    y_train
)

print("✅ Classification model fitted successfully!")


# ============================================================
# 12. PREDICT
# ============================================================

y_pred = classification_model.predict(
    X_test
)

print("✅ Predictions generated successfully!")


# ============================================================
# 13. EVALUATION METRICS
# ============================================================

accuracy = accuracy_score(
    y_test,
    y_pred
)

precision = precision_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)


print("\n" + "=" * 100)
print("CLASSIFICATION MODEL PERFORMANCE")
print("=" * 100)

print(f"Accuracy  : {accuracy:.4f}")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1-Score  : {f1:.4f}")


# ============================================================
# 14. CLASSIFICATION REPORT
# ============================================================

print("\n" + "=" * 100)
print("CLASSIFICATION REPORT")
print("=" * 100)

print(
    classification_report(
        y_test,
        y_pred,
        zero_division=0
    )
)


# ============================================================
# 15. EVALUATION METRIC DATAFRAME
# ============================================================

metrics_df = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1-Score"
    ],
    "Score": [
        accuracy,
        precision,
        recall,
        f1
    ]
})

print("\nEvaluation Metric Table:")
display(
    metrics_df.round(4)
)


# ============================================================
# 16. EVALUATION METRIC SCORE CHART
# ============================================================

plt.figure(figsize=(10, 6))

bars = plt.bar(
    metrics_df["Metric"],
    metrics_df["Score"]
)

plt.title(
    "Random Forest Classification - Evaluation Metric Scores"
)

plt.xlabel("Evaluation Metric")
plt.ylabel("Score")

# Add score labels
for bar, value in zip(
    bars,
    metrics_df["Score"]
):

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f"{value:.4f}",
        ha="center",
        va="bottom"
    )

plt.ylim(0, 1.05)
plt.tight_layout()
plt.show()


# ============================================================
# 17. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    y_pred
)

plt.figure(figsize=(8, 6))

ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=sorted(
        y_test.unique()
    )
).plot(
    ax=plt.gca(),
    xticks_rotation=45
)

plt.title(
    "Random Forest Classification - Confusion Matrix"
)

plt.tight_layout()
plt.show()


# ============================================================
# 18. 5-FOLD STRATIFIED CROSS-VALIDATION
# ============================================================

print("\n" + "=" * 100)
print("5-FOLD STRATIFIED CROSS-VALIDATION")
print("=" * 100)

skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

cv_results = cross_validate(
    classification_model,
    X,
    y,
    cv=skf,
    scoring={
        "Accuracy": "accuracy",
        "Precision": "precision_weighted",
        "Recall": "recall_weighted",
        "F1": "f1_weighted"
    },
    n_jobs=-1
)

cv_accuracy = cv_results[
    "test_Accuracy"
].mean()

cv_precision = cv_results[
    "test_Precision"
].mean()

cv_recall = cv_results[
    "test_Recall"
].mean()

cv_f1 = cv_results[
    "test_F1"
].mean()

print(f"CV Accuracy  : {cv_accuracy:.4f}")
print(f"CV Precision : {cv_precision:.4f}")
print(f"CV Recall    : {cv_recall:.4f}")
print(f"CV F1-Score  : {cv_f1:.4f}")


# ============================================================
# 19. CROSS-VALIDATION SCORE CHART
# ============================================================

cv_metrics_df = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1-Score"
    ],
    "CV Score": [
        cv_accuracy,
        cv_precision,
        cv_recall,
        cv_f1
    ]
})

plt.figure(figsize=(10, 6))

bars = plt.bar(
    cv_metrics_df["Metric"],
    cv_metrics_df["CV Score"]
)

plt.title(
    "Random Forest Classification - 5-Fold Cross-Validation"
)

plt.xlabel("Evaluation Metric")
plt.ylabel("Mean CV Score")

for bar, value in zip(
    bars,
    cv_metrics_df["CV Score"]
):

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f"{value:.4f}",
        ha="center",
        va="bottom"
    )

plt.ylim(0, 1.05)
plt.tight_layout()
plt.show()


# ============================================================
# 20. TEST VS CROSS-VALIDATION SUMMARY
# ============================================================

summary_df = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1-Score"
    ],

    "Test Score": [
        accuracy,
        precision,
        recall,
        f1
    ],

    "5-Fold CV Score": [
        cv_accuracy,
        cv_precision,
        cv_recall,
        cv_f1
    ]
})

print("\n" + "=" * 100)
print("TEST vs CROSS-VALIDATION PERFORMANCE")
print("=" * 100)

display(
    summary_df.round(4)
)


# ============================================================
# 21. SAVE MODEL
# ============================================================

model_file = os.path.join(
    model_path,
    "tourism_visitmode_random_forest_model.pkl"
)

joblib.dump(
    classification_model,
    model_file
)

print("\n✅ Classification model saved:")
print(model_file)


# ============================================================
# 22. SAVE EVALUATION RESULTS
# ============================================================

evaluation_file = os.path.join(
    model_path,
    "tourism_classification_evaluation.csv"
)

summary_df.to_csv(
    evaluation_file,
    index=False
)

print("\n✅ Evaluation results saved:")
print(evaluation_file)


# ============================================================
# 23. SAVE CLASS LABELS
# ============================================================

class_labels = sorted(
    y.unique()
)

labels_file = os.path.join(
    model_path,
    "tourism_visitmode_classes.pkl"
)

joblib.dump(
    class_labels,
    labels_file
)

print("\n✅ Class labels saved:")
print(labels_file)


print("\n" + "=" * 100)
print("✅ ML MODEL - 2 COMPLETED SUCCESSFULLY!")
print("=" * 100)# Visualizing evaluation Metric Score chart
````

#### 2. Cross- Validation & Hyperparameter Tuning

````python
# ============================================================
# ML MODEL - 2
# RANDOM FOREST CLASSIFICATION
# CROSS-VALIDATION + HYPERPARAMETER TUNING
# Target: VisitMode
# ============================================================

import os
import joblib
import pandas as pd
import numpy as np

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    GridSearchCV
)

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

print("=" * 90)
print("MODEL - 2: CROSS-VALIDATION & HYPERPARAMETER TUNING")
print("=" * 90)


# ============================================================
# 1. PATHS
# ============================================================

project_path = r"C:\Users\Admin\OneDrive\Desktop\TOURISM_PROJECT"

data_path = os.path.join(
    project_path,
    "DATA"
)

model_path = os.path.join(
    project_path,
    "models"
)

os.makedirs(
    model_path,
    exist_ok=True
)


# ============================================================
# 2. LOAD DATASETS
# ============================================================

transaction_df = pd.read_csv(
    os.path.join(data_path, "Transaction.csv"),
    encoding="latin1"
)

user_df = pd.read_csv(
    os.path.join(data_path, "User.csv"),
    encoding="latin1"
)

item_df = pd.read_csv(
    os.path.join(data_path, "Item.csv"),
    encoding="latin1"
)

type_df = pd.read_csv(
    os.path.join(data_path, "Type.csv"),
    encoding="latin1"
)

mode_df = pd.read_csv(
    os.path.join(data_path, "Mode.csv"),
    encoding="latin1"
)

print("✅ All datasets loaded.")


# ============================================================
# 3. MERGE DATASETS
# ============================================================

df = transaction_df.copy()

# User data
df = df.merge(
    user_df,
    on="UserId",
    how="left"
)

# Attraction data
item_clean = (
    item_df[
        [
            "AttractionId",
            "AttractionCityId",
            "AttractionTypeId"
        ]
    ]
    .drop_duplicates(
        subset=["AttractionId"]
    )
)

df = df.merge(
    item_clean,
    on="AttractionId",
    how="left"
)

# Attraction type
type_clean = (
    type_df[
        [
            "AttractionTypeId",
            "AttractionType"
        ]
    ]
    .drop_duplicates(
        subset=["AttractionTypeId"]
    )
)

df = df.merge(
    type_clean,
    on="AttractionTypeId",
    how="left"
)

# Visit mode name
mode_clean = (
    mode_df[
        [
            "VisitModeId",
            "VisitMode"
        ]
    ]
    .drop_duplicates(
        subset=["VisitModeId"]
    )
)

df = df.merge(
    mode_clean,
    left_on="VisitMode",
    right_on="VisitModeId",
    how="left",
    suffixes=("_id", "_name")
)

print("✅ Datasets merged.")
print("Merged shape:", df.shape)


# ============================================================
# 4. DEFINE TARGET
# ============================================================

target = "VisitMode_name"

if target not in df.columns:

    raise ValueError(
        "VisitMode target column was not found. "
        "Please check Mode.csv."
    )

df = df.dropna(
    subset=[target]
)

y = df[target].astype(str)


# ============================================================
# 5. SELECT FEATURES
# ============================================================

features = [
    "VisitYear",
    "VisitMonth",
    "ContinentId",
    "RegionId",
    "CountryId",
    "CityId",
    "AttractionCityId",
    "AttractionTypeId"
]

features = [
    col
    for col in features
    if col in df.columns
]

X = df[features].copy()

print("\nFeatures used:")
print(features)

print("\nTarget:")
print(target)


# ============================================================
# 6. TRAIN / TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining rows:", len(X_train))
print("Testing rows :", len(X_test))


# ============================================================
# 7. FEATURE TYPES
# ============================================================

numeric_features = X_train.select_dtypes(
    include=np.number
).columns.tolist()

categorical_features = [
    col
    for col in X_train.columns
    if col not in numeric_features
]

print("\nNumeric features:")
print(numeric_features)

print("\nCategorical features:")
print(categorical_features)


# ============================================================
# 8. PREPROCESSING
# ============================================================

numeric_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(
            strategy="median"
        )
    )
])

categorical_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(
            strategy="most_frequent"
        )
    ),
    (
        "encoder",
        OneHotEncoder(
            handle_unknown="ignore"
        )
    )
])

preprocessor = ColumnTransformer([
    (
        "numeric",
        numeric_pipeline,
        numeric_features
    ),
    (
        "categorical",
        categorical_pipeline,
        categorical_features
    )
])


# ============================================================
# 9. RANDOM FOREST CLASSIFIER
# ============================================================

rf_classifier = RandomForestClassifier(
    random_state=42,
    n_jobs=-1,
    class_weight="balanced"
)


# ============================================================
# 10. COMPLETE PIPELINE
# ============================================================

classification_model = Pipeline([
    (
        "preprocessor",
        preprocessor
    ),
    (
        "classifier",
        rf_classifier
    )
])


# ============================================================
# 11. STRATIFIED 5-FOLD CROSS-VALIDATION
# ============================================================

skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)


# ============================================================
# 12. HYPERPARAMETER GRID
# ============================================================

param_grid = {
    "classifier__n_estimators": [
        50,
        100
    ],

    "classifier__max_depth": [
        10,
        20
    ],

    "classifier__min_samples_split": [
        2,
        5
    ],

    "classifier__min_samples_leaf": [
        1,
        2
    ]
}

print("\n" + "=" * 90)
print("HYPERPARAMETER SEARCH")
print("=" * 90)

for parameter, values in param_grid.items():
    print(
        f"{parameter}: {values}"
    )


# ============================================================
# 13. GRID SEARCH
# ============================================================

grid_search = GridSearchCV(
    estimator=classification_model,
    param_grid=param_grid,
    scoring="f1_weighted",
    cv=skf,
    n_jobs=-1,
    verbose=1
)


# ============================================================
# 14. FIT GRID SEARCH
# ============================================================

print("\n" + "=" * 90)
print("STARTING GRID SEARCH")
print("=" * 90)

grid_search.fit(
    X_train,
    y_train
)

print(
    "\n✅ GRID SEARCH COMPLETED SUCCESSFULLY!"
)


# ============================================================
# 15. BEST PARAMETERS
# ============================================================

best_classification_model = (
    grid_search.best_estimator_
)

print("\n" + "=" * 90)
print("BEST PARAMETERS")
print("=" * 90)

print(
    grid_search.best_params_
)

print(
    "\nBest CV F1 Score:",
    round(
        grid_search.best_score_,
        4
    )
)


# ============================================================
# 16. TEST PREDICTION
# ============================================================

y_pred = (
    best_classification_model
    .predict(X_test)
)


# ============================================================
# 17. EVALUATION
# ============================================================

accuracy = accuracy_score(
    y_test,
    y_pred
)

precision = precision_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

print("\n" + "=" * 90)
print("TUNED MODEL TEST PERFORMANCE")
print("=" * 90)

print(
    f"Accuracy  : {accuracy:.4f}"
)

print(
    f"Precision : {precision:.4f}"
)

print(
    f"Recall    : {recall:.4f}"
)

print(
    f"F1-Score  : {f1:.4f}"
)


# ============================================================
# 18. CLASSIFICATION REPORT
# ============================================================

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred,
        zero_division=0
    )
)


# ============================================================
# 19. SAVE BEST MODEL
# ============================================================

model_file = os.path.join(
    model_path,
    "best_tourism_visitmode_deployment.joblib"
)

joblib.dump(
    best_classification_model,
    model_file,
    compress=3
)

print("\n✅ Best model saved:")
print(model_file)


# ============================================================
# 20. VERIFY SAVED MODEL
# ============================================================

loaded_model = joblib.load(
    model_file
)

print(
    "\n✅ Saved model loaded successfully!"
)


# ============================================================
# 21. SANITY CHECK
# ============================================================

sample_prediction = (
    loaded_model
    .predict(
        X_test.head(1)
    )
)

print(
    "Sample predicted VisitMode:",
    sample_prediction[0]
)


# ============================================================
# 22. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 90)
print("FINAL MODEL SUMMARY")
print("=" * 90)

print(
    "Model      : Random Forest Classifier"
)

print(
    "Validation : Stratified 5-Fold Cross-Validation"
)

print(
    "Optimization: GridSearchCV"
)

print(
    "Best CV F1 :",
    round(
        grid_search.best_score_,
        4
    )
)

print(
    "Test Accuracy:",
    round(
        accuracy,
        4
    )
)

print(
    "Test F1:",
    round(
        f1,
        4
    )
)

print("\n✅ MODEL 2 IS READY FOR DEPLOYMENT!")
print("=" * 90)
````

````python
# ============================================================
# ML MODEL - 3 IMPLEMENTATION
# COLLABORATIVE FILTERING RECOMMENDATION SYSTEM
#
# Tourism Experience Analytics Project
# ============================================================

import os
import joblib
import numpy as np
import pandas as pd

from scipy.sparse import csr_matrix
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error
)
from sklearn.model_selection import train_test_split
from sklearn.metrics.pairwise import cosine_similarity

print("=" * 100)
print("ML MODEL - 3: COLLABORATIVE FILTERING")
print("=" * 100)


# ============================================================
# 1. PATHS
# ============================================================

project_path = r"C:\Users\Admin\OneDrive\Desktop\TOURISM_PROJECT"
data_path = os.path.join(project_path, "DATA")
model_path = os.path.join(project_path, "models")

os.makedirs(model_path, exist_ok=True)


# ============================================================
# 2. LOAD REQUIRED DATASETS
# ============================================================

transaction_df = pd.read_csv(
    os.path.join(data_path, "Transaction.csv"),
    encoding="latin1"
)

item_df = pd.read_csv(
    os.path.join(data_path, "Item.csv"),
    encoding="latin1"
)


# ============================================================
# 3. PREPARE USER-ATTRACTION-RATING DATA
# ============================================================

recommendation_df = transaction_df[
    [
        "UserId",
        "AttractionId",
        "Rating"
    ]
].copy()

# Convert Rating to numeric
recommendation_df["Rating"] = pd.to_numeric(
    recommendation_df["Rating"],
    errors="coerce"
)

# Remove invalid records
recommendation_df = recommendation_df.dropna(
    subset=[
        "UserId",
        "AttractionId",
        "Rating"
    ]
)

# Remove duplicate user-attraction records by averaging ratings
recommendation_df = (
    recommendation_df
    .groupby(
        ["UserId", "AttractionId"],
        as_index=False
    )["Rating"]
    .mean()
)

print("\nPrepared interaction data:")
print(recommendation_df.shape)

print("\nSample interaction records:")
display(recommendation_df.head())


# ============================================================
# 4. TRAIN-TEST SPLIT
# ============================================================

train_data, test_data = train_test_split(
    recommendation_df,
    test_size=0.20,
    random_state=42
)

print("\n" + "=" * 100)
print("TRAIN-TEST SPLIT")
print("=" * 100)

print("Training interactions:", len(train_data))
print("Testing interactions :", len(test_data))


# ============================================================
# 5. CREATE USER AND ITEM INDEXES
# ============================================================

train_users = train_data["UserId"].unique()
train_items = train_data["AttractionId"].unique()

user_to_index = {
    user_id: index
    for index, user_id in enumerate(train_users)
}

item_to_index = {
    item_id: index
    for index, item_id in enumerate(train_items)
}

index_to_item = {
    index: item_id
    for item_id, index in item_to_index.items()
}


# ============================================================
# 6. CREATE USER-ITEM MATRIX
# ============================================================

row_indices = train_data["UserId"].map(
    user_to_index
)

column_indices = train_data["AttractionId"].map(
    item_to_index
)

rating_values = train_data["Rating"].astype(float)

user_item_matrix = csr_matrix(
    (
        rating_values,
        (
            row_indices,
            column_indices
        )
    ),
    shape=(
        len(user_to_index),
        len(item_to_index)
    )
)

print("\nUser-Item Matrix Shape:")
print(user_item_matrix.shape)


# ============================================================
# 7. FIT ITEM-ITEM COLLABORATIVE FILTERING
# ============================================================

print("\n" + "=" * 100)
print("FITTING COLLABORATIVE FILTERING MODEL")
print("=" * 100)

# Calculate similarity between attractions
item_similarity_matrix = cosine_similarity(
    user_item_matrix.T,
    dense_output=True
)

print(
    "Item Similarity Matrix Shape:",
    item_similarity_matrix.shape
)

print("✅ Collaborative Filtering model fitted successfully!")


# ============================================================
# 8. CREATE FALLBACK VALUES
# ============================================================

global_mean_rating = train_data["Rating"].mean()

item_mean_rating = (
    train_data
    .groupby("AttractionId")["Rating"]
    .mean()
    .to_dict()
)


# ============================================================
# 9. PREDICT RATING FUNCTION
# ============================================================

def predict_rating(
    user_id,
    attraction_id
):
    """
    Predict a user's rating for an attraction
    using item-item collaborative filtering.
    """

    # Unknown user
    if user_id not in user_to_index:

        return item_mean_rating.get(
            attraction_id,
            global_mean_rating
        )

    # Unknown attraction
    if attraction_id not in item_to_index:

        return item_mean_rating.get(
            attraction_id,
            global_mean_rating
        )

    user_index = user_to_index[user_id]
    target_item_index = item_to_index[
        attraction_id
    ]

    # User's existing ratings
    user_ratings = (
        user_item_matrix
        .getrow(user_index)
        .toarray()
        .flatten()
    )

    # Similarities to target attraction
    similarities = item_similarity_matrix[
        target_item_index
    ].copy()

    # Ignore similarity with itself
    similarities[target_item_index] = 0

    # Attractions already rated by user
    rated_items = np.where(
        user_ratings > 0
    )[0]

    if len(rated_items) == 0:

        return item_mean_rating.get(
            attraction_id,
            global_mean_rating
        )

    similarity_values = similarities[
        rated_items
    ]

    rating_values = user_ratings[
        rated_items
    ]

    denominator = np.sum(
        np.abs(similarity_values)
    )

    if denominator == 0:

        return item_mean_rating.get(
            attraction_id,
            global_mean_rating
        )

    prediction = (
        np.sum(
            similarity_values *
            rating_values
        )
        / denominator
    )

    return float(prediction)


# ============================================================
# 10. FIT + PREDICT ON TEST DATA
# ============================================================

print("\n" + "=" * 100)
print("PREDICTING ON TEST DATA")
print("=" * 100)

actual_ratings = []
predicted_ratings = []

for row in test_data.itertuples(
    index=False
):

    prediction = predict_rating(
        row.UserId,
        row.AttractionId
    )

    actual_ratings.append(
        row.Rating
    )

    predicted_ratings.append(
        prediction
    )


# ============================================================
# 11. EVALUATION
# ============================================================

if len(actual_ratings) > 0:

    mae = mean_absolute_error(
        actual_ratings,
        predicted_ratings
    )

    rmse = np.sqrt(
        mean_squared_error(
            actual_ratings,
            predicted_ratings
        )
    )

    print("\n" + "=" * 100)
    print("RECOMMENDATION MODEL PERFORMANCE")
    print("=" * 100)

    print(f"MAE  : {mae:.4f}")
    print(f"RMSE : {rmse:.4f}")

else:

    mae = np.nan
    rmse = np.nan

    print("⚠️ No valid test predictions generated.")


# ============================================================
# 12. ACTUAL vs PREDICTED RATINGS
# ============================================================

prediction_df = pd.DataFrame({
    "UserId": test_data["UserId"].values,
    "AttractionId": test_data["AttractionId"].values,
    "Actual Rating": actual_ratings,
    "Predicted Rating": predicted_ratings
})

print("\n" + "=" * 100)
print("ACTUAL vs PREDICTED RATINGS")
print("=" * 100)

display(
    prediction_df.head(20)
)


# ============================================================
# 13. RECOMMENDATION FUNCTION
# ============================================================

def recommend_attractions(
    user_id,
    top_n=10
):
    """
    Recommend top attractions for a user.
    """

    # Unknown user fallback
    if user_id not in user_to_index:

        popular_items = (
            train_data
            .groupby("AttractionId")
            .agg(
                AverageRating=("Rating", "mean"),
                NumberOfRatings=("Rating", "count")
            )
            .sort_values(
                ["AverageRating", "NumberOfRatings"],
                ascending=False
            )
            .head(top_n)
            .reset_index()
        )

        return popular_items

    user_index = user_to_index[user_id]

    # User's existing ratings
    user_ratings = (
        user_item_matrix
        .getrow(user_index)
        .toarray()
        .flatten()
    )

    rated_items = set(
        np.where(user_ratings > 0)[0]
    )

    candidate_items = [
        index
        for index in range(
            len(item_to_index)
        )
        if index not in rated_items
    ]

    recommendations = []

    for item_index in candidate_items:

        attraction_id = index_to_item[
            item_index
        ]

        predicted = predict_rating(
            user_id,
            attraction_id
        )

        recommendations.append({
            "AttractionId": attraction_id,
            "PredictedRating": predicted
        })

    recommendation_result = pd.DataFrame(
        recommendations
    )

    if recommendation_result.empty:
        return recommendation_result

    recommendation_result = (
        recommendation_result
        .sort_values(
            "PredictedRating",
            ascending=False
        )
        .head(top_n)
        .reset_index(drop=True)
    )

    return recommendation_result


# ============================================================
# 14. TEST RECOMMENDATION
# ============================================================

print("\n" + "=" * 100)
print("SAMPLE ATTRACTION RECOMMENDATIONS")
print("=" * 100)

sample_user = train_users[0]

print("Sample UserId:", sample_user)

sample_recommendations = recommend_attractions(
    sample_user,
    top_n=10
)

display(
    sample_recommendations
)


# ============================================================
# 15. ADD ATTRACTION NAMES
# ============================================================

attraction_lookup = (
    item_df[
        [
            "AttractionId",
            "Attraction",
            "AttractionAddress"
        ]
    ]
    .drop_duplicates(
        subset=["AttractionId"]
    )
)

if not sample_recommendations.empty:

    sample_recommendations = (
        sample_recommendations
        .merge(
            attraction_lookup,
            on="AttractionId",
            how="left"
        )
    )

    print("\nRecommendations with Attraction Details:")
    display(
        sample_recommendations
    )


# ============================================================
# 16. SAVE RECOMMENDATION MODEL
# ============================================================

recommendation_artifacts = {

    "user_to_index": user_to_index,

    "item_to_index": item_to_index,

    "index_to_item": index_to_item,

    "user_item_matrix": user_item_matrix,

    "item_similarity_matrix": item_similarity_matrix,

    "global_mean_rating": global_mean_rating,

    "item_mean_rating": item_mean_rating
}

recommendation_model_file = os.path.join(
    model_path,
    "tourism_recommendation_model.pkl"
)

joblib.dump(
    recommendation_artifacts,
    recommendation_model_file
)

print("\n✅ Recommendation model saved:")
print(recommendation_model_file)


# ============================================================
# 17. SAVE ATTRACTION DATA
# ============================================================

attraction_file = os.path.join(
    model_path,
    "tourism_attraction_data.csv"
)

attraction_lookup.to_csv(
    attraction_file,
    index=False
)

print("\n✅ Attraction data saved:")
print(attraction_file)


# ============================================================
# 18. SAVE MODEL METRICS
# ============================================================

metrics_df = pd.DataFrame({
    "Metric": [
        "MAE",
        "RMSE"
    ],
    "Score": [
        mae,
        rmse
    ]
})

metrics_file = os.path.join(
    model_path,
    "tourism_recommendation_metrics.csv"
)

metrics_df.to_csv(
    metrics_file,
    index=False
)

print("\n✅ Recommendation metrics saved:")
print(metrics_file)


print("\n" + "=" * 100)
print("✅ ML MODEL - 3 COMPLETED SUCCESSFULLY!")
print("=" * 100)# ML Model - 3 Implementation

# Fit the Algorithm

# Predict on the model
````

#### 2. Cross- Validation & Hyperparameter Tuning

````python
# ============================================================
# ML MODEL - 3
# CROSS-VALIDATION & HYPERPARAMETER TUNING
# Collaborative Filtering Recommendation System
# ============================================================

import os
import joblib
import numpy as np
import pandas as pd

from sklearn.model_selection import KFold
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error
)
from sklearn.metrics.pairwise import cosine_similarity
from scipy.sparse import csr_matrix

print("=" * 100)
print("ML MODEL - 3: CROSS-VALIDATION & HYPERPARAMETER TUNING")
print("COLLABORATIVE FILTERING")
print("=" * 100)


# ============================================================
# 1. LOAD TRANSACTION DATA
# ============================================================

project_path = r"C:\Users\Admin\OneDrive\Desktop\TOURISM_PROJECT"
data_path = os.path.join(project_path, "DATA")
model_path = os.path.join(project_path, "models")

os.makedirs(model_path, exist_ok=True)

transaction_df = pd.read_csv(
    os.path.join(data_path, "Transaction.csv"),
    encoding="latin1"
)

print("\nTransaction Dataset Shape:")
print(transaction_df.shape)


# ============================================================
# 2. PREPARE USER-ATTRACTION-RATING DATA
# ============================================================

rec_data = transaction_df[
    [
        "UserId",
        "AttractionId",
        "Rating"
    ]
].copy()

rec_data["Rating"] = pd.to_numeric(
    rec_data["Rating"],
    errors="coerce"
)

rec_data = rec_data.dropna(
    subset=[
        "UserId",
        "AttractionId",
        "Rating"
    ]
)

# Average repeated user-attraction ratings
rec_data = (
    rec_data
    .groupby(
        ["UserId", "AttractionId"],
        as_index=False
    )["Rating"]
    .mean()
)

print("\nUnique user-attraction interactions:")
print(len(rec_data))


# ============================================================
# 3. HYPERPARAMETERS TO TUNE
# ============================================================

k_values = [
    5,
    10,
    20,
    30
]

print("\nHyperparameters to test:")
print("Number of similar attractions (K):", k_values)


# ============================================================
# 4. FUNCTION FOR COLLABORATIVE FILTERING
# ============================================================

def build_recommendation_model(
    train_data
):
    """
    Build user-item matrix and item-item similarity matrix.
    """

    users = train_data["UserId"].unique()
    items = train_data["AttractionId"].unique()

    user_to_index = {
        user_id: i
        for i, user_id in enumerate(users)
    }

    item_to_index = {
        item_id: i
        for i, item_id in enumerate(items)
    }

    index_to_item = {
        i: item_id
        for item_id, i in item_to_index.items()
    }

    rows = train_data["UserId"].map(
        user_to_index
    )

    cols = train_data["AttractionId"].map(
        item_to_index
    )

    values = train_data["Rating"].astype(float)

    matrix = csr_matrix(
        (
            values,
            (rows, cols)
        ),
        shape=(
            len(users),
            len(items)
        )
    )

    similarity = cosine_similarity(
        matrix.T,
        dense_output=True
    )

    return {
        "user_to_index": user_to_index,
        "item_to_index": item_to_index,
        "index_to_item": index_to_item,
        "matrix": matrix,
        "similarity": similarity
    }


# ============================================================
# 5. RATING PREDICTION FUNCTION
# ============================================================

def predict_rating_cf(
    model,
    train_data,
    user_id,
    attraction_id,
    k=10
):
    """
    Predict a rating using top-K similar attractions.
    """

    global_mean = train_data["Rating"].mean()

    item_mean = (
        train_data
        .groupby("AttractionId")["Rating"]
        .mean()
        .to_dict()
    )

    user_to_index = model["user_to_index"]
    item_to_index = model["item_to_index"]
    matrix = model["matrix"]
    similarity = model["similarity"]

    # Unknown user
    if user_id not in user_to_index:
        return item_mean.get(
            attraction_id,
            global_mean
        )

    # Unknown attraction
    if attraction_id not in item_to_index:
        return item_mean.get(
            attraction_id,
            global_mean
        )

    user_index = user_to_index[user_id]
    target_index = item_to_index[attraction_id]

    # User ratings
    user_ratings = (
        matrix
        .getrow(user_index)
        .toarray()
        .ravel()
    )

    # Similarities to target attraction
    similarities = similarity[
        target_index
    ].copy()

    # Exclude target itself
    similarities[target_index] = 0

    rated_items = np.where(
        user_ratings > 0
    )[0]

    if len(rated_items) == 0:
        return item_mean.get(
            attraction_id,
            global_mean
        )

    # Similarities of items rated by the user
    rated_similarities = similarities[
        rated_items
    ]

    rated_values = user_ratings[
        rated_items
    ]

    # Keep only positive similarities
    positive_mask = (
        rated_similarities > 0
    )

    rated_similarities = (
        rated_similarities[
            positive_mask
        ]
    )

    rated_values = (
        rated_values[
            positive_mask
        ]
    )

    if len(rated_similarities) == 0:
        return item_mean.get(
            attraction_id,
            global_mean
        )

    # Select top-K similar items
    top_k = min(
        k,
        len(rated_similarities)
    )

    top_indices = np.argsort(
        rated_similarities
    )[-top_k:]

    top_similarities = (
        rated_similarities[
            top_indices
        ]
    )

    top_ratings = (
        rated_values[
            top_indices
        ]
    )

    denominator = np.sum(
        top_similarities
    )

    if denominator == 0:
        return item_mean.get(
            attraction_id,
            global_mean
        )

    prediction = np.sum(
        top_similarities *
        top_ratings
    ) / denominator

    return float(prediction)


# ============================================================
# 6. 5-FOLD CROSS-VALIDATION
# ============================================================

print("\n" + "=" * 100)
print("5-FOLD CROSS-VALIDATION")
print("=" * 100)

kf = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

cv_results = []

for k in k_values:

    fold_rmse = []
    fold_mae = []

    print(
        f"\nTesting K = {k}"
    )

    for fold, (
        train_index,
        validation_index
    ) in enumerate(
        kf.split(rec_data),
        start=1
    ):

        train_fold = rec_data.iloc[
            train_index
        ].copy()

        validation_fold = rec_data.iloc[
            validation_index
        ].copy()

        # --------------------------------------------
        # Fit recommendation model on training fold
        # --------------------------------------------

        fold_model = build_recommendation_model(
            train_fold
        )

        actual = []
        predicted = []

        # --------------------------------------------
        # Predict validation fold
        # --------------------------------------------

        for row in validation_fold.itertuples(
            index=False
        ):

            prediction = predict_rating_cf(
                fold_model,
                train_fold,
                row.UserId,
                row.AttractionId,
                k=k
            )

            actual.append(
                row.Rating
            )

            predicted.append(
                prediction
            )

        # --------------------------------------------
        # Fold metrics
        # --------------------------------------------

        if len(actual) > 0:

            fold_mae_value = (
                mean_absolute_error(
                    actual,
                    predicted
                )
            )

            fold_rmse_value = np.sqrt(
                mean_squared_error(
                    actual,
                    predicted
                )
            )

            fold_mae.append(
                fold_mae_value
            )

            fold_rmse.append(
                fold_rmse_value
            )

            print(
                f"Fold {fold} → "
                f"MAE: {fold_mae_value:.4f}, "
                f"RMSE: {fold_rmse_value:.4f}"
            )

    # --------------------------------------------
    # Average CV performance
    # --------------------------------------------

    mean_mae = np.mean(
        fold_mae
    )

    mean_rmse = np.mean(
        fold_rmse
    )

    cv_results.append({
        "K": k,
        "Mean CV MAE": mean_mae,
        "Mean CV RMSE": mean_rmse
    })

    print(
        f"K={k} → "
        f"Mean MAE: {mean_mae:.4f}, "
        f"Mean RMSE: {mean_rmse:.4f}"
    )


# ============================================================
# 7. DISPLAY HYPERPARAMETER RESULTS
# ============================================================

cv_results_df = pd.DataFrame(
    cv_results
)

print("\n" + "=" * 100)
print("HYPERPARAMETER TUNING RESULTS")
print("=" * 100)

display(
    cv_results_df.round(4)
)


# ============================================================
# 8. SELECT BEST HYPERPARAMETER
# ============================================================

best_result = (
    cv_results_df
    .sort_values(
        "Mean CV RMSE",
        ascending=True
    )
    .iloc[0]
)

best_k = int(
    best_result["K"]
)

best_cv_rmse = (
    best_result["Mean CV RMSE"]
)

best_cv_mae = (
    best_result["Mean CV MAE"]
)

print("\n" + "=" * 100)
print("BEST HYPERPARAMETER")
print("=" * 100)

print("Best K              :", best_k)
print(
    "Best CV MAE         :",
    f"{best_cv_mae:.4f}"
)
print(
    "Best CV RMSE        :",
    f"{best_cv_rmse:.4f}"
)


# ============================================================
# 9. FIT FINAL MODEL USING BEST K
# ============================================================

print("\n" + "=" * 100)
print("FITTING FINAL RECOMMENDATION MODEL")
print("=" * 100)

final_model = build_recommendation_model(
    rec_data
)

print("✅ Final collaborative filtering model fitted.")


# ============================================================
# 10. CREATE FINAL RECOMMENDATION FUNCTION
# ============================================================

def recommend_attractions_tuned(
    user_id,
    top_n=10
):
    """
    Generate top-N attraction recommendations
    using the tuned K value.
    """

    train_data = rec_data

    user_to_index = (
        final_model["user_to_index"]
    )

    item_to_index = (
        final_model["item_to_index"]
    )

    index_to_item = (
        final_model["index_to_item"]
    )

    matrix = final_model["matrix"]

    global_mean = (
        train_data["Rating"].mean()
    )

    item_mean = (
        train_data
        .groupby("AttractionId")["Rating"]
        .mean()
        .to_dict()
    )

    # --------------------------------------------------------
    # Unknown user → recommend popular/high-rated attractions
    # --------------------------------------------------------

    if user_id not in user_to_index:

        popular = (
            train_data
            .groupby("AttractionId")
            .agg(
                AverageRating=("Rating", "mean"),
                NumberOfRatings=("Rating", "count")
            )
            .sort_values(
                [
                    "AverageRating",
                    "NumberOfRatings"
                ],
                ascending=False
            )
            .head(top_n)
            .reset_index()
        )

        return popular


    # --------------------------------------------------------
    # Existing user
    # --------------------------------------------------------

    user_index = user_to_index[
        user_id
    ]

    user_ratings = (
        matrix
        .getrow(user_index)
        .toarray()
        .ravel()
    )

    rated_items = set(
        np.where(
            user_ratings > 0
        )[0]
    )

    candidate_items = [
        i
        for i in range(
            len(item_to_index)
        )
        if i not in rated_items
    ]

    recommendation_list = []

    for item_index in candidate_items:

        attraction_id = index_to_item[
            item_index
        ]

        predicted = predict_rating_cf(
            final_model,
            train_data,
            user_id,
            attraction_id,
            k=best_k
        )

        recommendation_list.append({
            "AttractionId": attraction_id,
            "PredictedRating": predicted
        })

    recommendations = pd.DataFrame(
        recommendation_list
    )

    if recommendations.empty:
        return recommendations

    recommendations = (
        recommendations
        .sort_values(
            "PredictedRating",
            ascending=False
        )
        .head(top_n)
        .reset_index(drop=True)
    )

    return recommendations


# ============================================================
# 11. TEST FINAL RECOMMENDATION MODEL
# ============================================================

print("\n" + "=" * 100)
print("SAMPLE RECOMMENDATIONS")
print("=" * 100)

sample_user = (
    rec_data["UserId"]
    .iloc[0]
)

print("Sample UserId:", sample_user)

sample_recommendations = (
    recommend_attractions_tuned(
        sample_user,
        top_n=10
    )
)

display(
    sample_recommendations
)


# ============================================================
# 12. SAVE CV RESULTS
# ============================================================

cv_file = os.path.join(
    model_path,
    "tourism_recommendation_cv_results.csv"
)

cv_results_df.to_csv(
    cv_file,
    index=False
)

print("\n✅ Cross-validation results saved:")
print(cv_file)


# ============================================================
# 13. SAVE BEST PARAMETERS
# ============================================================

parameters = {
    "best_k": best_k,
    "best_cv_mae": best_cv_mae,
    "best_cv_rmse": best_cv_rmse
}

parameter_file = os.path.join(
    model_path,
    "tourism_recommendation_best_parameters.pkl"
)

joblib.dump(
    parameters,
    parameter_file
)

print("\n✅ Best parameters saved:")
print(parameter_file)


# ============================================================
# 14. SAVE FINAL MODEL
# ============================================================

final_artifacts = {
    "best_k": best_k,
    "user_to_index": final_model["user_to_index"],
    "item_to_index": final_model["item_to_index"],
    "index_to_item": final_model["index_to_item"],
    "user_item_matrix": final_model["matrix"],
    "item_similarity_matrix": final_model["similarity"],
    "global_mean_rating": rec_data["Rating"].mean(),
    "item_mean_rating": (
        rec_data
        .groupby("AttractionId")["Rating"]
        .mean()
        .to_dict()
    )
}

recommendation_model_file = os.path.join(
    model_path,
    "tourism_recommendation_tuned_model.pkl"
)

joblib.dump(
    final_artifacts,
    recommendation_model_file
)

print("\n✅ Tuned recommendation model saved:")
print(recommendation_model_file)


# ============================================================
# 15. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 100)
print("FINAL RECOMMENDATION MODEL SUMMARY")
print("=" * 100)

print("Technique              : Item-Item Collaborative Filtering")
print("Cross-Validation       : 5-Fold K-Fold")
print("Hyperparameter tuned  : K (similar attractions)")
print("Best K                 :", best_k)
print(
    "Best CV MAE            :",
    f"{best_cv_mae:.4f}"
)
print(
    "Best CV RMSE           :",
    f"{best_cv_rmse:.4f}"
)

print("\n✅ CROSS-VALIDATION & HYPERPARAMETER TUNING COMPLETED!")
print("=" * 100)# ML Model - 3 Implementation with hyperparameter optimization techniques (i.e., GridSearch CV, RandomSearch CV, Bayesian Optimization etc.)

# Fit the Algorithm

# Predict on the model
````
