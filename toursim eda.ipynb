## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# ============================================================
# STEP 13: DATA WRANGLING
# Tourism Experience Analytics Project
# ============================================================

import pandas as pd

print("=" * 80)
print("DATA WRANGLING")
print("=" * 80)

# ------------------------------------------------------------
# 1. Create copies of original datasets
# ------------------------------------------------------------

transaction = transaction_df.copy()
user = user_df.copy()
city = city_df.copy()
country = country_df.copy()
region = region_df.copy()
continent = continent_df.copy()
item = item_df.copy()
type_df_w = type_df.copy()
mode = mode_df.copy()


# ------------------------------------------------------------
# 2. Merge Transaction with User
# Relationship: Transaction.UserId → User.UserId
# ------------------------------------------------------------

merged_df = transaction.merge(
    user,
    on="UserId",
    how="left",
    suffixes=("", "_User")
)

print(f"After Transaction + User : {merged_df.shape}")


# ------------------------------------------------------------
# 3. Merge User information with City
# Relationship: User.CityId → City.CityId
# ------------------------------------------------------------

merged_df = merged_df.merge(
    city,
    on="CityId",
    how="left",
    suffixes=("", "_City")
)

print(f"After + City             : {merged_df.shape}")


# ------------------------------------------------------------
# 4. Merge City with Country
# Relationship: City.CountryId → Country.CountryId
# ------------------------------------------------------------

merged_df = merged_df.merge(
    country,
    on="CountryId",
    how="left",
    suffixes=("", "_Country")
)

print(f"After + Country          : {merged_df.shape}")


# ------------------------------------------------------------
# 5. Merge Country with Region
# Relationship: Country.RegionId → Region.RegionId
# ------------------------------------------------------------

merged_df = merged_df.merge(
    region,
    on="RegionId",
    how="left",
    suffixes=("", "_Region")
)

print(f"After + Region           : {merged_df.shape}")


# ------------------------------------------------------------
# 6. Merge Region with Continent
# Relationship: Region.ContinentId → Continent.ContinentId
# ------------------------------------------------------------

merged_df = merged_df.merge(
    continent,
    on="ContinentId",
    how="left",
    suffixes=("", "_Continent")
)

print(f"After + Continent        : {merged_df.shape}")


# ------------------------------------------------------------
# 7. Merge Transaction with Item/Attraction
# Relationship: Transaction.AttractionId → Item.AttractionId
# ------------------------------------------------------------

merged_df = merged_df.merge(
    item,
    on="AttractionId",
    how="left",
    suffixes=("", "_Item")
)

print(f"After + Item             : {merged_df.shape}")


# ------------------------------------------------------------
# 8. Merge Attraction with Type
# Relationship: Item.AttractionTypeId → Type.AttractionTypeId
# ------------------------------------------------------------

merged_df = merged_df.merge(
    type_df_w,
    on="AttractionTypeId",
    how="left",
    suffixes=("", "_Type")
)

print(f"After + Attraction Type   : {merged_df.shape}")


# ------------------------------------------------------------
# 9. Merge Transaction with Visit Mode
# Relationship: Transaction.VisitMode → Mode.VisitModeId
# ------------------------------------------------------------

merged_df = merged_df.merge(
    mode,
    left_on="VisitMode",
    right_on="VisitModeId",
    how="left",
    suffixes=("", "_Mode")
)

print(f"After + Visit Mode        : {merged_df.shape}")


# ------------------------------------------------------------
# 10. Remove completely duplicate rows
# ------------------------------------------------------------

before_duplicates = len(merged_df)

merged_df = merged_df.drop_duplicates().reset_index(drop=True)

after_duplicates = len(merged_df)

print("\nDuplicate rows removed:", before_duplicates - after_duplicates)


# ------------------------------------------------------------
# 11. Display final dataset information
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("FINAL WRANGLED DATASET")
print("=" * 80)

print("Rows    :", merged_df.shape[0])
print("Columns :", merged_df.shape[1])

print("\nFirst 5 Rows:")
display(merged_df.head())


# ------------------------------------------------------------
# 12. Check missing values after merging
# ------------------------------------------------------------

print("\nMissing Values After Data Wrangling:")

missing_after_merge = merged_df.isnull().sum()

missing_after_merge = missing_after_merge[
    missing_after_merge > 0
].sort_values(ascending=False)

if missing_after_merge.empty:
    print("✅ No missing values found after merging.")
else:
    display(missing_after_merge.to_frame("Missing Values"))


print("\n" + "=" * 80)
print("✅ DATA WRANGLING COMPLETED SUCCESSFULLY!")
print("=" * 80)# Write your code to make your dataset analysis ready.
````

#### Chart - 1

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ------------------------------------------------------------
# 1. Create a safe analysis dataframe
# ------------------------------------------------------------

# Start from Transaction data
viz_df = transaction_df.copy()

# Merge User information
viz_df = viz_df.merge(
    user_df,
    on="UserId",
    how="left"
)

# Merge Mode information
viz_df = viz_df.merge(
    mode_df,
    left_on="VisitMode",
    right_on="VisitModeId",
    how="left"
)

# Merge Attraction information
# Only use non-null unique AttractionId records
item_clean = item_df.dropna(subset=["AttractionId"]).drop_duplicates(
    subset=["AttractionId"]
)

viz_df = viz_df.merge(
    item_clean,
    on="AttractionId",
    how="left"
)

# Merge Attraction Type
viz_df = viz_df.merge(
    type_df,
    on="AttractionTypeId",
    how="left"
)

# Merge City
city_clean = city_df.drop_duplicates(subset=["CityId"])

viz_df = viz_df.merge(
    city_clean[["CityId", "CityName"]],
    on="CityId",
    how="left"
)

# Merge Country
country_clean = country_df.drop_duplicates(subset=["CountryId"])

viz_df = viz_df.merge(
    country_clean[["CountryId", "Country"]],
    on="CountryId",
    how="left"
)

# Merge Region
region_clean = (
    region_df.dropna(subset=["RegionId"])
             .drop_duplicates(subset=["RegionId"])
)

viz_df = viz_df.merge(
    region_clean[["RegionId", "Region"]],
    on="RegionId",
    how="left"
)

# Merge Continent
continent_clean = (
    continent_df.dropna(subset=["ContinentId"])
                .drop_duplicates(subset=["ContinentId"])
)

viz_df = viz_df.merge(
    continent_clean[["ContinentId", "Continent"]],
    on="ContinentId",
    how="left"
)

# Remove accidental duplicate rows after merging
viz_df = viz_df.drop_duplicates().reset_index(drop=True)

# Remove placeholder "-" values where useful
for col in ["VisitMode", "VisitMode_y", "AttractionType", "Attraction",
            "CityName", "Country", "Region", "Continent"]:
    if col in viz_df.columns:
        viz_df[col] = viz_df[col].replace("-", np.nan)

print("✅ Visualization dataset created successfully!")
print("Shape:", viz_df.shape)


# ============================================================
# CHART 1: Rating Distribution
# ============================================================

plt.figure(figsize=(10, 5))
sns.histplot(viz_df["Rating"].dropna(), bins=5, kde=True)
plt.title("1. Distribution of Attraction Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Visits")
plt.tight_layout()
plt.show()# Chart - 1 visualization code
````

````python
# ============================================================
# CHART 2: VISIT MODE COUNT
# Tourism Experience Analytics Project
# ============================================================

import os
import pandas as pd
import matplotlib.pyplot as plt

print("=" * 80)
print("CHART 2: NUMBER OF VISITS BY VISIT MODE")
print("=" * 80)


# ------------------------------------------------------------
# 1. Dataset Path
# ------------------------------------------------------------

data_path = r"C:\Users\Admin\OneDrive\Desktop\TOURISM_PROJECT\DATA"


# ------------------------------------------------------------
# 2. Load datasets
# ------------------------------------------------------------

transaction_df = pd.read_csv(
    os.path.join(data_path, "Transaction.csv"),
    encoding="latin1"
)

mode_df = pd.read_csv(
    os.path.join(data_path, "Mode.csv"),
    encoding="latin1"
)


# ------------------------------------------------------------
# 3. Prepare Mode table
# ------------------------------------------------------------

mode_clean = mode_df[
    ["VisitModeId", "VisitMode"]
].drop_duplicates(
    subset=["VisitModeId"]
)


# ------------------------------------------------------------
# 4. Merge Transaction with Mode
# ------------------------------------------------------------

chart_df = transaction_df.merge(
    mode_clean,
    left_on="VisitMode",
    right_on="VisitModeId",
    how="left",
    suffixes=("_Transaction", "_Mode")
)


# ------------------------------------------------------------
# 5. Find the correct VisitMode name column
# ------------------------------------------------------------

print("\nColumns after merge:")
print(chart_df.columns.tolist())

if "VisitMode_Mode" in chart_df.columns:
    mode_column = "VisitMode_Mode"

elif "VisitMode_y" in chart_df.columns:
    mode_column = "VisitMode_y"

elif "VisitMode" in chart_df.columns:
    mode_column = "VisitMode"

else:
    raise KeyError(
        "VisitMode name column was not found after merging. "
        "Please check the columns printed above."
    )

print("\nUsing column:", mode_column)


# ------------------------------------------------------------
# 6. Count Visit Modes
# ------------------------------------------------------------

mode_counts = (
    chart_df[mode_column]
    .dropna()
    .value_counts()
)


# ------------------------------------------------------------
# 7. Display counts
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("VISIT MODE COUNTS")
print("=" * 80)

print(mode_counts)


# ------------------------------------------------------------
# 8. Create chart
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

bars = plt.bar(
    mode_counts.index.astype(str),
    mode_counts.values
)

plt.title(
    "Number of Visits by Visit Mode"
)

plt.xlabel(
    "Visit Mode"
)

plt.ylabel(
    "Number of Visits"
)

plt.xticks(
    rotation=45,
    ha="right"
)


# Add values above bars
for bar, value in zip(
    bars,
    mode_counts.values
):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        str(value),
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()


print("\n✅ CHART 2 COMPLETED SUCCESSFULLY!")
````

#### Chart - 3

````python
# ============================================================
# CHART 3: AVERAGE RATING BY VISIT MODE
# Tourism Experience Analytics Project
# ============================================================

import os
import pandas as pd
import matplotlib.pyplot as plt

print("=" * 80)
print("CHART 3: AVERAGE RATING BY VISIT MODE")
print("=" * 80)


# ------------------------------------------------------------
# 1. Dataset Path
# ------------------------------------------------------------

data_path = r"C:\Users\Admin\OneDrive\Desktop\TOURISM_PROJECT\DATA"


# ------------------------------------------------------------
# 2. Load required datasets
# ------------------------------------------------------------

transaction_df = pd.read_csv(
    os.path.join(data_path, "Transaction.csv"),
    encoding="latin1"
)

mode_df = pd.read_csv(
    os.path.join(data_path, "Mode.csv"),
    encoding="latin1"
)


# ------------------------------------------------------------
# 3. Clean Mode dataset
# ------------------------------------------------------------

mode_clean = mode_df[
    ["VisitModeId", "VisitMode"]
].drop_duplicates(
    subset=["VisitModeId"]
)


# ------------------------------------------------------------
# 4. Merge Transaction + Mode
# ------------------------------------------------------------

chart_df = transaction_df.merge(
    mode_clean,
    left_on="VisitMode",
    right_on="VisitModeId",
    how="left",
    suffixes=("_Transaction", "_Mode")
)


# ------------------------------------------------------------
# 5. Identify the correct VisitMode column
# ------------------------------------------------------------

if "VisitMode_Mode" in chart_df.columns:

    mode_col = "VisitMode_Mode"

elif "VisitMode_y" in chart_df.columns:

    mode_col = "VisitMode_y"

elif "VisitMode_Transaction" in chart_df.columns:

    mode_col = "VisitMode_Transaction"

elif "VisitMode" in chart_df.columns:

    mode_col = "VisitMode"

else:

    raise KeyError(
        "VisitMode column was not found after merging."
    )


print("Visit Mode column used:", mode_col)


# ------------------------------------------------------------
# 6. Convert Rating to numeric
# ------------------------------------------------------------

chart_df["Rating"] = pd.to_numeric(
    chart_df["Rating"],
    errors="coerce"
)


# ------------------------------------------------------------
# 7. Calculate average rating by Visit Mode
# ------------------------------------------------------------

avg_mode_rating = (
    chart_df
    .dropna(subset=[mode_col, "Rating"])
    .groupby(mode_col)["Rating"]
    .mean()
    .sort_values(ascending=False)
)


# ------------------------------------------------------------
# 8. Display results
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("AVERAGE RATING BY VISIT MODE")
print("=" * 80)

print(
    avg_mode_rating.round(2)
)


# ------------------------------------------------------------
# 9. Create bar chart
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

bars = plt.bar(
    avg_mode_rating.index.astype(str),
    avg_mode_rating.values
)

plt.title(
    "Average Attraction Rating by Visit Mode"
)

plt.xlabel(
    "Visit Mode"
)

plt.ylabel(
    "Average Rating"
)

plt.xticks(
    rotation=45,
    ha="right"
)


# ------------------------------------------------------------
# 10. Add values above bars
# ------------------------------------------------------------

for bar, value in zip(
    bars,
    avg_mode_rating.values
):

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f"{value:.2f}",
        ha="center",
        va="bottom"
    )


plt.tight_layout()
plt.show()


print("\n✅ CHART 3 COMPLETED SUCCESSFULLY!")
````

Comparing average ratings across categories makes differences in satisfaction easy to understand.

Some visitor groups may provide higher average ratings than others.

#### Chart - 15 - Pair Plot

````python
# ============================================================
# PAIR PLOT - TOURISM EXPERIENCE ANALYTICS
# Complete standalone code
# ============================================================

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. Dataset path
# ------------------------------------------------------------

data_path = r"C:\Users\Admin\OneDrive\Desktop\TOURISM_PROJECT\DATA"


# ------------------------------------------------------------
# 2. Load required datasets
# ------------------------------------------------------------

transaction_df = pd.read_csv(
    data_path + r"\Transaction.csv",
    encoding="latin1"
)

user_df = pd.read_csv(
    data_path + r"\User.csv",
    encoding="latin1"
)


# ------------------------------------------------------------
# 3. Create analysis dataset
# ------------------------------------------------------------

viz_df = transaction_df.merge(
    user_df,
    on="UserId",
    how="left"
)


# ------------------------------------------------------------
# 4. Select numerical variables for Pair Plot
# ------------------------------------------------------------

pair_cols = [
    "VisitYear",
    "VisitMonth",
    "Rating"
]

# Keep only columns available in the dataset
pair_cols = [
    col for col in pair_cols
    if col in viz_df.columns
]


# ------------------------------------------------------------
# 5. Prepare data
# ------------------------------------------------------------

pair_data = viz_df[pair_cols].copy()

# Convert variables to numeric
for col in pair_cols:
    pair_data[col] = pd.to_numeric(
        pair_data[col],
        errors="coerce"
    )

# Remove missing values
pair_data = pair_data.dropna()


# ------------------------------------------------------------
# 6. Take sample for faster plotting
# ------------------------------------------------------------

sample_size = min(1000, len(pair_data))

pair_data = pair_data.sample(
    n=sample_size,
    random_state=42
)


# ------------------------------------------------------------
# 7. Display information
# ------------------------------------------------------------

print("=" * 70)
print("PAIR PLOT DATA")
print("=" * 70)

print("Variables used:", pair_cols)
print("Rows used:", len(pair_data))

print("=" * 70)


# ------------------------------------------------------------
# 8. Create Pair Plot
# ------------------------------------------------------------

sns.pairplot(
    pair_data,
    diag_kind="hist"
)

plt.suptitle(
    "Pair Plot - Tourism Experience Analytics",
    y=1.02
)

plt.show()


print("✅ PAIR PLOT COMPLETED SUCCESSFULLY!")
````

##### 1. Why did you pick the specific chart?

. We selected a pair plot because it allows us to visualize relationships between multiple numerical variables simultaneously. 
. It also shows the distribution of individual variables on the diagonal.

##### 2. What is/are the insight(s) found from the chart?

. The pair plot helps identify whether numerical variables have positive, negative, or weak relationships.
. It also helps identify clusters, unusual observations, and the overall distribution of variables. 
. In our Tourism project, it can help examine the relationship between visit timing and attraction ratings.
